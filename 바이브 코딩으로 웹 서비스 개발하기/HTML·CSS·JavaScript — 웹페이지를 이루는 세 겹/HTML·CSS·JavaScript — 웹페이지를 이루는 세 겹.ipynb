## AI는 잘 만드는데, 내 말은 왜 안 통할까

---

웹페이지는 하나의 덩어리처럼 보이지만, 안을 들여다보면 성격이 다른 세 가지가 겹쳐 있습니다. 화면에 **무엇이** 있는지(구조), 그것이 **어떻게 보이는지**(표현), 사용자가 무언가 했을 때 **무엇이 바뀌는지**(동작)입니다. 개발자들은 이 셋을 각각 HTML·CSS·JavaScript로 나누어 다룹니다.

![웹 브라우저가 그려낸 웹페이지 화면](https://commons.wikimedia.org/wiki/Special:FilePath/WorldWideWeb.1.png?width=640)

*웹 브라우저가 HTML 문서를 읽어 화면으로 그려낸 모습(초기 브라우저 WorldWideWeb). 우리가 "웹페이지"라 부르는 것은 이렇게 브라우저가 그려낸 결과입니다. 이미지: WorldWideWeb, 퍼블릭 도메인, via Wikimedia Commons.*

오늘은 이 세 겹의 문법을 깊이 파지는 않습니다. 대신 웹페이지를 세 겹으로 나누어 보는 연습을 합니다. 이렇게 보면 AI가 만든 화면에서 고칠 곳을 더 정확히 짚고, 결과가 기대와 다를 때 어느 쪽을 다시 봐야 할지 판단하기 좋습니다. 오늘은 먼저 이 눈을 만들고, 그다음 그 눈으로 직접 무언가를 만들며 확인합니다.

먼저 이름과 생김새부터 풀어두겠습니다. 문법을 외울 필요는 없습니다. "아, 이렇게 생겼구나" 정도만 보고 넘어가면 됩니다.

**HTML**(HyperText Markup Language)은 화면에 **무엇이** 담기는지를 적는 문서입니다. 제목, 문단, 버튼 같은 정보에 "이건 제목", "이건 버튼"이라고 의미를 붙입니다.

```html
<!-- 태그로 의미를 붙입니다. h1 = 가장 큰 제목, button = 누르는 버튼 -->
<h1>내 첫 웹페이지</h1>
<button>시작하기</button>
```

**CSS**(Cascading Style Sheets)는 그 정보가 **어떻게 보일지**를 정하는 규칙입니다. 색, 간격, 크기, 배치가 여기서 정해집니다.

```css
/* "h1 제목은 이렇게 보여라" — 파란색으로, 가운데 정렬로 */
h1 {
  color: blue;
  text-align: center;
}
```

**JavaScript**는 사용자가 **무언가를 했을 때 무엇이 바뀔지**를 정하는 프로그래밍 언어입니다. 버튼을 누르면 화면이 달라지는 반응이 여기서 만들어집니다.

```javascript
// 버튼을 누르면(click) 인사말 창을 띄웁니다
button.addEventListener('click', () => {
  alert('안녕하세요!');
});
```

세 이름이 따로 존재하는 데는 이유가 있습니다. 셋은 같은 날 한 사람이 설계한 세트가 아니라, 웹이 자라면서 서로 다른 문제를 풀기 위해 서로 다른 시기에 태어난 도구들입니다. 그 내력은 섹션 2에서 살펴봅니다. 지금은 "세 겹은 각자 맡은 일이 다르다"는 감각만 챙기면 충분합니다.

### 🎯 오늘의 핵심 한 문장

> **웹페이지는 구조(HTML)·표현(CSS)·동작(JS) 세 겹이고, 셋을 나눠 AI에게 지시하면 원하는 화면을 빠르게 만들고 고칠 수 있습니다.**

| 겹 | 담당하는 일 | AI에게 시킬 때의 말 | 결과 판단 질문 |
|---|---|---|---|
| HTML | 화면에 무엇이 있는지 정합니다 | "제목, 안내 카드, 버튼을 의미에 맞게 배치해줘" | 필요한 정보와 업무 용어가 빠지지 않았는가 |
| CSS | 그것이 어떻게 보이는지 정합니다 | "모바일에서도 읽히게 간격, 색, 글자 크기를 조정해줘" | 눈이 어디로 가는지, 깨지는 구간은 없는가 |
| JavaScript | DOM의 클래스·속성·상태를 바꿔 사용자가 했을 때 무엇이 바뀌는지 정합니다 | "버튼을 누르면 화면의 내용이 바뀌게 해줘" | 클릭·입력·상태 변화가 기대대로 일어나는가 |

이 세 겹은 개발자가 코드를 정리하려고 만든 구분이기도 하지만, AI에게 일을 맡기는 사람에게도 쓸모가 큽니다. 요청을 세 겹으로 나누면 AI가 고쳐야 할 지점이 좁아지고, 여러분도 결과를 감으로만 보지 않게 됩니다.

## 세 겹을 나누어 이해하기 — 이름에 담긴 사연

---

오늘 필요한 문법은 많지 않습니다. 대신 각 겹이 **왜 따로 존재하는지**를 알아두면, AI 출력에서 어디를 봐야 하는지가 분명해집니다. 세 겹은 각각 다른 시기에, 다른 문제를 풀기 위해 태어났고, 이름 안에 그 사연이 들어 있습니다. 사연을 따라가며 이름을 풀어보겠습니다.

### HTML — 잃어버린 문서에서 시작된 이야기

1980년대 말 유럽입자물리연구소(CERN)는 세계에서 손꼽히게 큰 연구소였고, 그만큼 정보가 엉켜 있는 곳이었습니다. 수천 명의 연구자가 저마다 다른 컴퓨터, 다른 프로그램으로 문서를 만들었고, 연구자가 연구소를 떠나면 그 사람 머릿속에만 있던 자료의 위치도 함께 사라졌습니다. "그 실험 결과, 어느 문서에 있죠?"라는 질문에 답하려면 결국 사람을 찾아다녀야 했습니다.

이곳의 연구원 **팀 버너스리(Tim Berners-Lee)** 가 1989년에 낸 제안서의 아이디어는 이랬습니다. 문서를 컴퓨터별로 쌓아두지 말고, 문서 속 단어에서 관련된 다른 문서로 **바로 건너뛸 수 있게 연결하자.** 읽던 순서를 뛰어넘어(hyper) 이동하는 텍스트라는 뜻에서, 이런 연결 방식을 **하이퍼텍스트(hypertext)** 라고 부릅니다. 당시 그의 상사는 제안서 표지에 "막연하지만 흥미롭다(Vague but exciting...)"라고 적었습니다. 그 막연한 제안이 월드 와이드 웹이 되었습니다.

제안이 승인되자 버너스리는 1990년 말, NeXT사의 컴퓨터 한 대로 최초의 웹 브라우저와 최초의 웹 서버를 모두 만들었습니다. 세계 최초의 웹 서버는 연구실 책상 위의 이 검은 컴퓨터였는데, 누가 실수로 끌까 봐 본체에 손글씨 스티커가 붙어 있었습니다. "이 기계는 서버입니다. 전원을 끄지 마시오!!" 웹의 첫 인프라는 데이터센터가 아니라 스티커 한 장으로 지켜진 셈입니다.

![NeXTcube — 세계 최초의 웹 서버](https://commons.wikimedia.org/wiki/Special:FilePath/First_Web_Server.jpg?width=640)

*팀 버너스리가 CERN에서 쓴 세계 최초의 웹 서버(NeXTcube). "이 기계는 서버입니다. 전원을 끄지 마시오"라는 손글씨 스티커가 붙어 있습니다. 사진: Coolcaesar, CC BY-SA 3.0, via Wikimedia Commons.*

1991년에 공개된 세계 최초의 웹사이트는 CERN이 복원해 두어 지금도 접속할 수 있습니다.

> 🔗 **직접 가보세요**: http://info.cern.ch — 최초의 웹사이트. 배경도 색도 버튼도 없이 링크 달린 글자뿐인 화면, 즉 "CSS도 JavaScript도 없던 웹"이 어떤 모습인지 확인할 수 있습니다.

한 가지 결정이 이 발명의 운명을 갈랐습니다. 1993년 4월 CERN은 웹 기술 전체를 누구나 로열티 없이 쓸 수 있게 공개했습니다. 특허를 걸고 사용료를 받는 길도 있었지만 그러지 않았고, 그 덕분에 웹은 특정 회사의 제품이 아니라 모두의 기반이 되었습니다. 여러분이 오늘 HTML을 허락도 비용도 없이 쓰는 이유가 이 결정에 있습니다.

덧붙이면 '문서를 연결한다'는 발상 자체는 버너스리의 발명이 아닙니다. 1945년 공학자 버니바 부시(Vannevar Bush)가 "As We May Think"라는 글에서 연관 자료를 연결해 따라가는 기계(Memex)를 상상했고, 1965년 테드 넬슨(Ted Nelson)이 그런 글쓰기에 하이퍼텍스트라는 이름을 붙였습니다. 버너스리의 공헌은 이 오래된 상상을 인터넷 위에서 실제로 작동하게 만든 것입니다.

HTML이라는 이름은 여기서 그대로 나옵니다. **H**yper**T**ext **M**arkup **L**anguage, "하이퍼텍스트를 적기 위한 마크업 언어"입니다. '마크업'은 편집자가 원고 여백에 "여기는 제목, 여기는 인용"이라고 표시하던 교정 관행에서 온 말입니다. HTML 태그도 같은 일을 합니다. `<h1>`은 "이것은 문서에서 가장 높은 수준의 제목", `<button>`은 "이것은 누르는 버튼"이라고 정보의 **의미**를 표시합니다. JavaScript가 DOM의 클래스·속성·상태를 바꾸면 CSS가 이를 표현하고 HTML의 `hidden` 같은 속성이 함께 반응을 완성합니다.

태생이 이렇다 보니 HTML을 검토하는 기준도 꾸밈이 아니라 정보입니다. 필요한 정보가 다 있는가, 각 정보에 맞는 의미가 붙었는가. 예를 들어 병원 원무 안내 페이지를 AI에게 맡겼더니 '진료 안내', '예약', '오시는 길' 카드가 나왔다고 합시다. 겉보기에는 번듯하지만, 실제 창구에서 하루에도 수십 번 답하는 '진단서·제증명 발급' 안내가 빠져 있습니다. 이 누락은 코딩 지식이 아니라 현장을 아는 눈에만 보입니다. 그래서 이 겹의 최종 검토자는 여러분입니다.

> 📚 웹이 태어나는 이 과정이 더 궁금하시면, W3C가 정리한 "A short history of the Web"(https://www.w3.org/History.html)에서 이어 읽으실 수 있습니다.

### CSS — "제목 색을 전부 바꿔주세요"의 악몽

초기의 웹에는 디자인이라는 개념이 거의 없었습니다. 문서의 모양은 브라우저가 정해주는 대로였고, 글쓴이가 색이나 크기를 바꾸고 싶으면 꾸밈 태그를 본문 곳곳에 직접 끼워 넣어야 했습니다. 페이지가 몇 개일 때는 버틸 만했습니다. 그런데 문서가 수백 개로 늘어난 사이트에서 "제목 색을 전부 바꿔주세요"라는 요청이 오면, 담당자는 모든 문서를 하나하나 열어 고쳐야 했습니다. 내용과 꾸밈이 한 덩어리로 붙어 있었기 때문입니다.

그 시절의 개인 홈페이지를 상상해 보면 이렇습니다. 형광 초록 글자에 우주 별 배경 무늬, 반짝이는 GIF 아이콘, 페이지마다 제각각인 글꼴, 그리고 하단의 방문자 카운터. 만든 사람은 밤새 꾸밈 태그를 본문에 끼워 넣었고, 일주일 뒤 "전체 분위기를 차분하게 바꾸고 싶다"는 마음이 들면 그 태그를 전부 다시 걷어내야 했습니다.

1994년, CERN에서 버너스리와 함께 일하던 **호콘 비움 리(Håkon Wium Lie)** 는 인쇄 세계에서 힌트를 가져옵니다. 신문과 잡지는 기사 원고와 지면의 조판 규칙을 따로 관리합니다. 기사를 한 글자도 고치지 않고 조판 규칙만 바꾸면 지면 전체의 인상이 달라집니다. 웹도 그렇게 하자 — 꾸밈 규칙을 문서 밖의 **스타일시트(style sheet)** 로 분리하자는 제안이었고, 1996년 표준이 되었습니다.

이름의 첫 단어 **캐스케이딩(cascading)** 은 '폭포처럼 층층이 흘러내린다'는 뜻입니다. 화면 하나에는 여러 스타일 규칙이 동시에 적용될 수 있습니다. 브라우저의 기본 규칙, 사이트 공통 규칙, 페이지 개별 규칙. 규칙들이 겹칠 때 어느 쪽이 이기는지, 우선순위가 폭포처럼 층을 이루며 정해져 있다는 뜻이 이름에 담겨 있습니다. 그래서 CSS는 **C**ascading **S**tyle **S**heets입니다.

내용과 꾸밈의 분리가 실제로 얼마나 강력한지 보여준 상징적인 실험이 있습니다. 2003년 디자이너 데이브 셰이(Dave Shea)가 만든 CSS Zen Garden은 HTML 파일 하나를 고정해 두고, 전 세계 디자이너들이 CSS 파일만 바꿔 제출하게 했습니다. 같은 HTML인데 어떤 CSS를 입히느냐에 따라 일본 정원이 되기도 하고 잡지 표지가 되기도 합니다. 내용은 한 글자도 바뀌지 않았는데 말입니다.

> 🔗 **직접 가보세요**: https://csszengarden.com — 디자인 목록을 눌러 같은 HTML이 몇 번이고 다른 옷을 입는 것을 확인해 보세요. 실습에서 "내용은 그대로 두고 표현만 바꿔줘"라고 요청할 때, 여러분은 이 원리를 쓰는 것입니다.

이 내력이 CSS의 검토 기준을 알려줍니다. CSS가 담당하는 것은 정보가 아니라 **읽히는 방식**입니다. 눈이 어디로 먼저 가는가, 모바일 폭에서 깨지는 구간은 없는가. AI에게는 "예쁘게"보다 "모바일 390px에서도 카드가 한 열로 읽히고, 주요 버튼이 한눈에 보이게"처럼 관찰 가능한 조건을 주는 편이 낫습니다.

> 📚 CSS가 태어난 배경이 더 궁금하시면, 제안자가 직접 정리한 W3C의 "A brief history of CSS until 2016"(https://www.w3.org/Style/CSS20/history.html)을 보실 수 있습니다.

### CSS를 말로 고치기 위한 최소 지도 — 박스모델

### 학생이 자연어로 말할 수 있는 단어장

CSS 속성명을 암기해서 지시할 필요는 없습니다. 아래처럼 **눈에 보이는 문제를 일상어로 말하면 AI가 CSS 속성으로 번역**합니다. 다만 "어느 요소"와 "모바일에서도" 같은 조건을 함께 말하면 결과를 확인하기 좋습니다.

| 이렇게 말해보세요 | AI가 주로 살펴볼 CSS 힌트 | 더 구체적인 요청 예시 |
|---|---|---|
| "카드 바깥에 여백이 필요해" | `margin` | "카드와 카드 사이가 답답해요. 모바일에서도 바깥 간격을 조금 늘려줘." |
| "버튼 안의 글자가 테두리에 너무 붙어 있어" | `padding` | "버튼 글자와 테두리 사이의 안쪽 여백을 늘려서 누르기 편하게 해줘." |
| "버튼들이 서로 너무 붙어 있어" | `gap`, `margin` | "버튼을 가로로 놓고 버튼 사이 간격을 일정하게 벌려줘." |
| "카드에 테두리를 치고 싶어" | `border` | "각 카드에 연한 회색 테두리와 둥근 모서리를 넣어줘." |
| "내용을 가운데로 모아줘" | `display: flex`, `justify-content`, `margin: auto` | "카드 안의 버튼은 가운데 정렬하고, 페이지 본문은 너무 넓지 않게 가운데에 놓아줘." |

> 💬 AI에게는 이렇게 요청해보세요: **"카드 내용은 바꾸지 말고, 카드 바깥 간격과 버튼 사이 간격을 모바일에서도 읽기 좋게 조정해줘. 어떤 CSS를 바꿨는지도 알려줘."**
>
> 📚 박스모델의 content·padding·border·margin 구조와 `box-sizing` 동작은 [MDN CSS box model](https://developer.mozilla.org/en-US/docs/Learn_web_development/Core/Styling_basics/Box_model)에서 확인할 수 있습니다.

### JavaScript — 열흘 만에 태어난 언어

1995년의 웹페이지는 잘 차려진 인쇄물에 가까웠습니다. 정보는 담겨 있지만, 눌러도 아무 일도 일어나지 않는 종이였습니다. 입력한 값이 잘못됐는지 확인하는 일조차 서버까지 갔다 와야 알 수 있었습니다.

당시 브라우저 시장을 이끌던 넷스케이프(Netscape)는 마이크로소프트가 곧 이 시장에 뛰어든다는 압박 속에서, 페이지가 사용자의 행동에 **그 자리에서 반응**하게 만들 언어를 원했습니다. 이 일을 맡은 **브렌던 아이크(Brendan Eich)** 는 첫 버전을 열흘 남짓한 기간에 만들었습니다. 이름은 Mocha에서 LiveScript로, 다시 JavaScript로 바뀌었는데, 마지막 개명은 당시 최고 화제였던 언어 Java의 인기에 올라탄 마케팅이었습니다. 두 언어는 이름만 닮은 남남이라, 개발자들 사이에는 "자바와 자바스크립트는 햄과 햄스터의 관계"라는 농담이 오래 전해집니다.

뒷이야기가 하나 더 있습니다. 아이크는 원래 학술 언어 Scheme을 브라우저에 넣자는 이야기를 듣고 넷스케이프에 합류했습니다. 그런데 경영진의 요구는 "문법은 Java를 닮게"였습니다. 그래서 JavaScript는 겉모습은 Java를 닮았지만 속은 전혀 다른 언어가 되었고, 이 어색한 동거가 이름 문제와 함께 오해의 뿌리가 되었습니다.

이 언어의 첫 임무는 소박했습니다. **폼 검증** — 입력칸을 확인하는 일입니다. 1995년의 인터넷은 전화선 모뎀으로 연결됐고, 페이지 하나를 다시 받는 데 수십 초가 걸렸습니다. 회원가입 폼에서 이메일 한 칸을 잘못 쓰면, 그 느린 왕복 끝에 "다시 입력하세요" 한 줄이 돌아왔습니다. 제출 버튼을 누르기 전에 브라우저가 그 자리에서 "이메일 형식이 아닙니다"라고 말해 주는 것 — JavaScript가 처음으로 세상에 쓸모를 증명한 장면입니다.

JavaScript가 맡는 것은 시간에 따라 변하는 것들입니다. 클릭, 입력, 필터, 토글처럼 사용자가 무언가를 한 뒤에 화면이 달라지는 일이 전부 이 겹에서 일어납니다. 오늘은 복잡한 알고리즘을 다루지 않습니다. 버튼을 누르면 목록이 바뀌는 정도의 동작으로 충분합니다. 검토 기준은 하나입니다. "무엇을 누르면 무엇이 바뀌어야 하는가"가 기대대로 일어나는가.

이 작은 언어가 웹의 주인공이 되는 결정적 장면은 10년 뒤에 옵니다. 2005년 구글 지도가 등장했을 때, 사람들은 지도를 마우스로 잡아 끌었습니다. 그때까지의 웹 지도는 화살표 버튼을 누르고 페이지 전체가 다시 로드되기를 기다리는 물건이었는데, 구글 지도는 끄는 동안에도 새 지도 조각을 뒤에서 계속 받아와 이어 붙였습니다. 페이지를 새로 고치지 않고 서버와 데이터를 주고받는 이 기법은 곧 AJAX라는 이름을 얻었고(제시 제임스 개럿의 2005년 글에서 명명), 웹페이지가 '문서'에서 '애플리케이션'으로 넘어가는 분기점이 되었습니다.

그런데 여기서 궁금증이 하나 생깁니다. 지도를 끄는 손을 따라가면서, 동시에 뒤에서 데이터를 받는 일이 어떻게 가능할까요? JavaScript는 분명 한 번에 한 가지 일을 하는 언어인데 말입니다. 이 질문은 잠시 뒤 **[오늘의 CS]** 에서 답합니다.

> 📚 급하게 태어난 언어가 웹의 표준이 되어가는 과정이 궁금하시면, 만든 당사자가 공저한 역사 논문 "JavaScript: The First 20 Years"(https://dl.acm.org/doi/10.1145/3386327)에서 이어 읽으실 수 있습니다.

### 세 겹의 내력이 지시의 언어가 됩니다

정리하면, 웹은 처음부터 세 겹으로 설계된 것이 아니라 30년에 걸쳐 문제를 하나씩 풀며 지금의 모양이 되었습니다. 잃어버린 문서를 연결하는 뼈대, 내용과 꾸밈의 분리, 행동에 대한 반응. 각 겹이 태어난 이유가 다르니 맡은 일과 검토 기준도 다릅니다.

| 겹 | 태어난 문제 | 검토 기준 |
|---|---|---|
| HTML | 흩어진 정보를 의미 있게 연결하기 | 정보와 용어가 빠지지 않았는가 |
| CSS | 내용과 꾸밈이 붙어 있어 수정이 번지는 문제 | 읽히는가, 깨지는 폭은 없는가 |
| JavaScript | 눌러도 반응하지 않는 정적인 문서 | 행동 뒤의 변화가 기대대로인가 |

AI에게 지시할 때 이 구분을 쓰는 이유도 같습니다. "페이지를 고쳐줘"는 세 겹 중 어디를 고칠지 정하지 않은 요청이고, "카드 정보는 그대로 두고 모바일 간격만 조정해줘"는 겹을 지정한 요청입니다. 겹을 지정하면 AI가 손댈 범위가 좁아지고, 이미 괜찮은 부분이 덜 흔들립니다.

### CS 교양 — '관심사의 분리'라는 오래된 원칙

내용·표현·동작을 나누는 발상은 웹만의 것이 아닙니다. 컴퓨터 과학자 **에츠허르 데이크스트라(Edsger W. Dijkstra)** 는 1974년의 글에서, 복잡한 문제를 다룰 때는 한 번에 한 측면에만 집중하고 나머지는 잠시 접어두는 태도를 **관심사의 분리(separation of concerns)** 라고 불렀습니다. 큰 문제를 통째로 붙들면 어디서부터 손대야 할지 알기 어렵지만, 측면을 나누면 측면마다 판단 기준이 생깁니다.

웹의 세 겹은 이 원칙이 화면 설계에 적용된 결과이고, 협업 구조에도 그대로 이어집니다. 콘텐츠 담당자는 문구와 정보 구조를 보고, 디자이너는 표현을 보고, 개발자는 동작을 봅니다. 서로 다른 직군이 같은 페이지를 두고 각자의 겹을 고칠 수 있는 것은 겹이 나뉘어 있기 때문입니다. AI와 일할 때도 같습니다. 여러분이 관심사를 나눠 말하면, AI도 한 번에 한 측면에 집중합니다.

> 🔍 **판단해보세요**: 모바일에서 버튼 두 개가 겹치는 문제를 발견했습니다. AI가 작업하기 더 좋은 요청은 어느 쪽일까요?
>
> A. "버튼이 겹치니까 `.nav-button`에 `flex-wrap: wrap`이랑 `min-width: 120px`를 줘."
> B. "모바일에서 버튼 두 개가 겹쳐서 누르기 어려워. 버튼 안 글자는 그대로 두고, 버튼이 줄바꿈되게 고쳐줘."
>
> 왜 그런지 한 줄로 적어보세요.

<details>
<summary>💡 검토 포인트</summary>

B를 권합니다. A가 더 구체적으로 보이지만, 해법(CSS 속성명)을 사람이 지정한 요청입니다. 속성을 모르는 채 추측한 명세는 틀릴 수 있고, 틀려도 AI는 그 지시를 따릅니다. B는 관찰(겹침)과 기대(줄바꿈), 그리고 지킬 것(글자 내용)을 나눠 말했습니다 — 고칠 관심사와 건드리지 말 관심사를 분리한 것입니다. 해법을 고르는 일은 AI의 몫으로 남겨두면 됩니다.

**[CS 교양]** 프론트엔드와 백엔드 - 웹은 왜 두 쪽으로 나뉘었을까

지금까지 본 HTML, CSS, JavaScript는 사용자가 브라우저에서 직접 만나는 영역입니다. 화면에 어떤 정보가 놓이는지, 어떤 색과 간격으로 읽히는지, 버튼을 눌렀을 때 무엇이 바뀌는지를 다룹니다. 개발자들은 이 보이는 쪽을 보통 **프론트엔드(frontend)** 라고 부릅니다.

웹서비스에는 보이지 않는 쪽도 있습니다. 사용자가 제출한 신청서를 저장하고, 로그인한 사람만 자기 기록을 보게 하고, 여러 사람이 같은 데이터를 보더라도 기준이 흔들리지 않게 관리하는 영역입니다. 이 화면 뒤의 처리 영역을 **백엔드(backend)** 라고 부릅니다.

식당으로 비유하면 프론트엔드는 손님이 앉는 홀과 메뉴판, 주문 화면에 가깝습니다. 손님은 메뉴를 읽고, 주문을 고르고, 직원에게 요청합니다. 백엔드는 주방과 창고, 주문 관리 장부에 가깝습니다. 재료가 남아 있는지 확인하고, 주문 순서를 맞추고, 결제 기록과 예약 기록을 보관합니다. 홀만 있어도 메뉴판은 보이지만 음식이 나오지 않습니다. 주방만 있어도 손님은 무엇을 주문해야 할지 알기 어렵습니다. 웹서비스도 두 쪽이 함께 움직일 때 사용자 경험이 이어집니다.

| 구분 | 프론트엔드 | 백엔드 |
|---|---|---|
| 사용자가 만나는 위치 | 브라우저 화면, 버튼, 입력창, 카드, 목록 | 서버, API, 데이터 저장소, 권한 검사 |
| 오늘 배운 연결 | HTML, CSS, JavaScript가 담당하는 보이는 쪽 | 뒤 강의에서 다룰 서버와 데이터 처리 |
| 주요 질문 | 사용자가 무엇을 보고, 누르고, 입력하는가 | 입력된 데이터가 어디에 저장되고, 누가 다시 볼 수 있는가 |
| 식당 비유 | 홀, 메뉴판, 주문 화면 | 주방, 창고, 주문 장부 |
| AI에게 줄 말 | 화면 구조, 배치, 클릭 후 변화 | 저장 규칙, API 응답, 권한, 오류 처리 |

웹이 처음부터 이렇게 나뉜 것은 아닙니다. 초기 웹페이지는 문서를 보여주는 역할이 컸고, 화면과 서버의 일이 지금보다 덜 복잡했습니다. 시간이 지나며 로그인, 결제, 검색, 추천, 실시간 알림처럼 사용자의 행동과 데이터가 늘어났습니다. 화면은 더 많은 상호작용을 맡게 되었고, 서버는 저장과 보안, 여러 사용자 사이의 일관성을 맡게 되었습니다. 한쪽에 모든 일을 몰아두면 수정이 어렵고 위험도 커졌기 때문에, 웹은 점차 보이는 쪽과 처리하는 쪽을 나누어 설계하게 되었습니다.

바이브 코딩에서도 이 구분은 중요합니다. 오늘처럼 화면을 만들 때는 AI에게 프론트엔드 기준으로 말해야 합니다. 예를 들어 "업무 카드가 모바일에서 한 열로 보이게 해 주세요"는 프론트엔드 요청입니다. 반대로 "상담 기록을 저장하고, 작성자만 수정하게 해 주세요"는 백엔드 요청입니다. 두 요청을 섞어 던지면 AI가 화면은 만들었지만 저장이 빠지거나, 저장 코드는 만들었지만 사용자가 눌러볼 화면이 부족한 결과가 나올 수 있습니다.

오늘은 프론트엔드의 세 겹을 먼저 익힙니다. 백엔드와 API는 뒤 강의에서 다룹니다. 그때 오늘의 식당 비유를 다시 떠올리면 좋습니다. 프론트엔드는 손님이 주문을 이해하게 돕는 자리이고, 백엔드는 주문이 실제 기록과 처리로 이어지게 만드는 자리입니다.

## [오늘의 CS] 동기·비동기, 프로세스·스레드·코어

## [오늘의 CS] 동기·비동기, 프로세스·스레드·코어 - 웹이 멈추지 않는 이유

---

앞에서 남겨둔 질문을 다시 꺼냅니다. 구글 지도는 어떻게 손을 따라 움직이면서 동시에 데이터를 받아올까요? 유튜브는 영상을 재생하면서 어떻게 그 아래 댓글을 계속 불러올까요? 답의 열쇠는 컴퓨터가 **기다림**을 다루는 방식에 있습니다.

### 동기와 비동기 - 전화와 문자

**동기(synchronous)** 는 전화입니다. 상대가 받을 때까지 기다리고, 통화가 끝날 때까지 붙잡혀 있습니다. 그동안 다른 일은 못 합니다. **비동기(asynchronous)** 는 문자입니다. 보내놓고 하던 일을 계속하다가, 답장 알림이 오면 그때 확인합니다.

프로그래밍에서도 같습니다. 동기 방식은 한 작업이 끝날 때까지 다음 줄로 넘어가지 않는 것이고, 비동기 방식은 오래 걸리는 작업을 걸어두고 다음 일을 진행하다가, 끝났다는 신호가 오면 그 결과를 처리하는 것입니다.

카페 진동벨이 좋은 그림입니다. 진동벨이 없다면 주문한 사람은 커피가 나올 때까지 카운터 앞에 서 있어야 하고(동기), 뒷사람들 줄은 계속 길어집니다. 진동벨을 받으면 자리로 가서 하던 일을 하다가, 벨이 울리면 받으러 갑니다(비동기). 커피가 나오는 시간 자체는 같습니다. 달라지는 것은 **기다리는 동안 다른 일을 할 수 있느냐**입니다.

### JavaScript는 일꾼이 한 명입니다

이 구분이 JavaScript에서 특히 중요한 이유가 있습니다. 웹페이지 안의 JavaScript는 기본적으로 **한 번에 한 가지 일**만 합니다. 일꾼이 한 명이라는 뜻으로, 이를 **싱글 스레드(single thread)** 로 동작한다고 말합니다.

일꾼이 한 명인데 동기 방식으로만 일한다면 어떻게 될까요? 버튼 클릭을 처리하는 코드가 서버 응답을 3초간 기다리며 서 있으면, 그 3초 동안 화면의 모든 것이 멈춥니다. 스크롤도 안 되고, 다른 버튼도 안 눌리고, 재생되던 영상도 로딩 아이콘도 멈춰버립니다. 어떤 사이트에서 버튼을 누른 뒤 화면 전체가 굳어버린 경험이 있다면, 그중 일부가 바로 이 문제입니다.

그래서 JavaScript는 오래 걸리는 일을 비동기로 처리합니다. 서버에 데이터를 요청할 때(`fetch`), 시간을 재놓을 때(`setTimeout`), 요청만 걸어두고 즉시 다음 일로 넘어갑니다. 무거운 일은 브라우저가 뒤에서 대신 처리하고, 끝나면 결과를 대기줄에 올려놓습니다. 한 명뿐인 일꾼은 지금 하던 일을 마친 뒤 대기줄을 확인해 결과를 처리합니다. 이 대기줄을 돌리는 구조를 **이벤트 루프(event loop)** 라고 부릅니다(MDN 이벤트 루프 문서: https://developer.mozilla.org/en-US/docs/Web/JavaScript/Event_loop).

이 비동기 요청에는 출생의 비밀이 하나 있습니다. 페이지를 새로 고치지 않고 서버와 데이터를 주고받는 핵심 부품(XMLHttpRequest)은 1990년대 말 마이크로소프트의 아웃룩 웹메일 팀이, 웹메일을 데스크톱 프로그램처럼 만들려고 개발한 것입니다. 넷스케이프가 열흘 만에 만든 언어와, 그 경쟁사가 웹메일용으로 만든 부품이 만나 몇 년 뒤 구글 지도에서 꽃을 피운 셈입니다. 웹의 역사는 이렇게 한 회사의 큰 그림이 아니라, 서로 다른 사건들의 우연한 합류로 흘러왔습니다.

AI가 만든 코드에서 `async`, `await`, `setTimeout`, `.then(...)` 같은 단어를 만나면, 그것이 바로 이 진동벨 패턴입니다. 외울 필요는 없습니다. "여기는 기다리지 않고 걸어두는 부분이구나"라고 읽을 수 있으면 오늘은 충분합니다.

### 프로세스와 스레드 - 작업실과 일꾼

그런데 브라우저는 어떻게 JavaScript의 일을 '뒤에서 대신' 처리해 줄까요? 시야를 한 층 아래, 운영체제로 내립니다. Day 4에서 커널이 자원을 중개한다고 배웠는데, 그 자원을 배분받는 단위가 프로세스와 스레드입니다.

**프로세스(process)** 는 실행 중인 프로그램 하나입니다. 각 프로세스는 독립된 작업실을 씁니다 — 자기만의 메모리를 갖고, 다른 프로세스의 메모리를 넘볼 수 없습니다. **스레드(thread)** 는 그 작업실 안에서 실제로 일하는 일꾼입니다. 한 작업실에 일꾼이 여러 명일 수 있고, 일꾼들은 작업실의 재료(메모리)를 공유합니다.

왜 이런 구분이 필요한지는 2008년 크롬의 탄생이 잘 보여줍니다. 그 시절 브라우저는 대부분 프로그램 전체가 프로세스 하나였습니다. 웹이 문서이던 시절에는 그래도 괜찮았습니다 — 탭이 죽으면 다시 열면 그만이었으니까요. 그런데 AJAX 이후 웹이 애플리케이션이 되자 사정이 달라졌습니다. 탭 하나의 오류가 브라우저 전체를 끌고 넘어지면, 쓰던 웹메일 초안과 열어 둔 문서 열 개가 한꺼번에 사라졌습니다. 잃을 것이 많아진 것입니다. 구글은 크롬을 내놓으면서 '탭마다 독립된 프로세스'라는 운영체제식 격리를 선택했고, 이 설계를 일반 사용자에게 설명하려고 만화가 스콧 매클라우드(Scott McCloud)에게 소개 만화를 그리게 한 일화도 남아 있습니다(설계 문서: https://www.chromium.org/developers/design-documents/multi-process-architecture/).

그 결과가 지금 여러분이 매일 보는 화면입니다. 크롬에서 탭 하나가 "페이지가 응답하지 않습니다"라며 죽어도 다른 탭들은 멀쩡합니다. 한 작업실에 불이 나도 옆 작업실은 안전합니다. 활동 상태 보기(macOS)나 작업 관리자(Windows)를 열면 크롬 항목이 수십 개 떠 있는 것을 직접 볼 수 있습니다 — 탭과 확장 프로그램마다 작업실이 따로라는 증거입니다.

일을 여러 갈래로 나누는 방법도 두 가지로 갈립니다.

| | 멀티스레딩 | 멀티프로세싱 |
|---|---|---|
| 그림 | 한 작업실에 일꾼 여러 명 | 작업실 자체를 여러 개 |
| 메모리 | 공유합니다 | 각자 독립입니다 |
| 강점 | 가볍고, 일꾼끼리 재료를 바로 주고받습니다 | 하나가 죽어도 나머지는 안전합니다 |
| 약점 | 같은 재료를 동시에 건드리면 꼬일 수 있습니다 | 무겁고, 작업실 사이 재료 전달에 비용이 듭니다 |
| 눈앞의 예 | 브라우저 한 탭 안의 렌더링·네트워크 일꾼들 | 크롬의 탭들, 앞으로 만날 개발 서버와 빌드 도구 |

멀티스레딩의 약점 한 줄은 기억해 둘 가치가 있습니다. 일꾼 두 명이 같은 장부에 동시에 숫자를 쓰면 기록이 꼬입니다. 그래서 여러 일꾼이 재료를 공유하는 프로그램은 강력하지만 조심스럽게 설계해야 하고, JavaScript가 아예 일꾼을 한 명으로 정한 것도 이 골치 아픈 문제를 피하는 선택이었습니다.

### 코어 — 일손은 결국 몇 개인가

마지막 층은 하드웨어입니다. **코어(core)** 는 CPU 안에서 실제로 계산하는 물리적인 일손입니다. 코어가 8개면 진짜로 동시에 8가지 일을 할 수 있습니다.

![여러 개의 코어가 든 CPU 칩](https://commons.wikimedia.org/wiki/Special:FilePath/Intel_CPU_Core_i7_2600K_Sandy_Bridge_top.jpg?width=640)

*한 칩 안에 여러 개의 코어가 담긴 CPU(Intel Core i7). 이 물리적 일손들이 수많은 프로세스·스레드를 나눠 맡습니다. 사진: Eric Gaba (Sting), CC BY-SA 3.0, via Wikimedia Commons.*

코어가 여러 개가 된 데에는 사연이 있습니다. 2000년대 초까지 CPU 경쟁은 속도(클럭) 경쟁이었습니다. 작년에 산 컴퓨터보다 올해 컴퓨터가 그냥 더 빨랐고, 소프트웨어는 가만히 있어도 해마다 빨라졌습니다. 한때 10GHz 시대가 온다는 전망까지 나왔지만, 이 경쟁은 물리학에 막힙니다. 클럭을 올릴수록 발열과 전력 소모가 감당할 수 없이 커진 것입니다. 칩 회사들은 방향을 틀었습니다 — 하나를 더 빠르게 만들 수 없다면, 적당히 빠른 것을 여러 개 넣자. 그렇게 멀티코어 시대가 왔습니다. 소프트웨어 쪽에서는 허브 서터(Herb Sutter)가 2005년 "공짜 점심은 끝났다(The Free Lunch Is Over)"라는 글로 이 전환을 요약했습니다. 이제 프로그램은 가만히 있으면 빨라지지 않고, 여러 일손을 쓰도록 설계해야 빨라진다는 선언이었습니다. 동시성이 일부 전문가의 관심사에서 모든 소프트웨어의 조건이 된 것이 이때부터입니다. 그런데 여러분의 컴퓨터에는 지금도 수백 개의 프로세스와 그보다 많은 스레드가 있습니다. 일손 8개가 어떻게 그 많은 일을 할까요?

운영체제가 아주 빠르게 번갈아 실행합니다. 한 스레드를 잠깐 실행하고, 멈추고, 다음 스레드로 — 이 전환이 워낙 빨라서 사람 눈에는 모든 것이 동시에 도는 것처럼 보입니다. 혼자서 여러 테이블을 도는 능숙한 홀 직원처럼, 코어는 수많은 스레드 사이를 오갑니다.

이 번갈아 쓰기, 즉 **시분할(time-sharing)** 의 뿌리는 1960년대로 거슬러 갑니다. 당시 컴퓨터는 방 하나 크기에 기관에 한 대뿐이라, 연구자들은 천공카드 뭉치로 프로그램을 제출하고 결과를 몇 시간 뒤에 받았습니다. 쉼표 하나 틀리면 그날 하루가 사라졌습니다. MIT의 존 매카시(John McCarthy)는 컴퓨터가 여러 사람 사이를 아주 빠르게 오가며 각자가 "자기 컴퓨터를 독차지한 듯" 느끼게 하자고 제안했고, 페르난도 코르바토(Fernando Corbató) 팀이 CTSS라는 시스템으로 구현했습니다. 덤으로 남은 유산이 하나 있습니다. 여러 사람이 한 컴퓨터를 나눠 쓰게 되자 서로의 파일을 구분할 장치가 필요해졌고, 그렇게 태어난 것이 컴퓨터 **비밀번호**입니다. 오늘 아침 여러분이 입력한 비밀번호는 시분할의 부산물입니다.

> 🔍 **내 컴퓨터로 확인해보세요** (Day 4의 터미널을 다시 엽니다)
>
> - macOS: `sysctl -n hw.ncpu` — 내 컴퓨터의 코어(논리 코어) 수가 숫자 하나로 나옵니다.
> - Windows PowerShell: `echo $env:NUMBER_OF_PROCESSORS`
>
> 그 숫자와, 활동 상태 보기/작업 관리자에 떠 있는 프로세스 수를 비교해 보세요. 일손보다 일감이 몇 배는 많은데도 화면이 멈추지 않는 것 — 그것이 운영체제가 하는 일입니다.

### 다시 세 겹으로 — 이 개념이 여러분의 지시를 바꿉니다

정리하면 이렇게 연결됩니다. 코어는 물리적 일손이고, 운영체제는 그 일손을 프로세스와 스레드에 나눠 주고, 브라우저는 탭마다 프로세스를 열어 그 안에서 여러 스레드로 일하며, 그중 JavaScript 일꾼은 한 명이라 오래 걸리는 일을 비동기로 걸어둡니다. 구글 지도가 손을 따라오면서 데이터를 받는 것도, 영상을 보면서 동시에 댓글을 입력할 수 있는 것도 이 구조 위에서 일어나는 일입니다.

이 개념은 코드를 쓰기 위한 것이 아니라 **문제를 말하기 위한** 것입니다. 비교해 보세요.

- "게임이 가끔 버벅여요." → AI가 원인을 추측합니다.
- "목록 항목이 많아지면 화면이 뚝뚝 끊겨. 무거운 계산이 화면 갱신을 막는 것 같은데, 계산을 나누거나 뒤로 미룰 수 있어?" → AI가 살펴볼 범위가 좁아집니다.

### 🔧 1차 연습 — "테트리스 만들어줘"

이번에는 터미널 기반의 개발 에이전트가 아니라 일반적인 **웹 채팅형 AI**(claude.ai 또는 ChatGPT 등의 브라우저 채팅 창)에서 요청합니다. 새 대화를 열고 이렇게만 입력합니다.

```text
테트리스 만들어줘.
```

정말 이게 전부입니다. 잠시 뒤 채팅 옆에 게임 미리보기가 뜨고, 그 자리에서 바로 플레이할 수 있습니다. 대개 방향키 조작과 줄 삭제까지 갖춘 판이 나옵니다. 방향키가 먹지 않으면 게임 화면을 한 번 클릭해서 선택한 뒤 다시 눌러 보세요. 도구에 따라 미리보기 없이 코드만 보여줄 수도 있습니다 — 그 경우 "바로 실행해서 볼 수 있게 해줘"라고 요청하거나, 미리보기를 지원하는 도구로 시도해 보세요. 안 되는 부분이 있다면 실망할 일이 아니라, 그게 바로 첫 재요청의 재료입니다.

한 줄 요청이 통하는 이유가 있습니다. 테트리스는 1984년부터 전 세계가 공유해 온 게임이라, 블록 모양·회전·줄 삭제·게임오버 같은 규칙이 이미 AI 안에 들어 있습니다. **"테트리스"라는 이름 자체가 규칙 명세서** 역할을 하는 것입니다. 요청이 통하고 안 통하고는 길이가 아니라 담긴 정보량이 가릅니다. "테트리스"는 짧지만 규칙이 가득 담긴 한 줄인 셈입니다. 뒤에서 만들 '내가 정한 게임'에는 이런 공용 이름이 없어서, 그때는 규칙을 직접 적어 정보를 채워야 합니다.

#### 플레이가 아니라 관찰까지가 연습입니다

미리보기 화면에서 **코드 보기**로 전환해 세 겹을 눈으로 확인합니다. 코드를 다 이해할 필요는 없습니다. "어느 겹이 어디에 있는가"만 찾습니다.

| 겹 | 찾아볼 것 |
|---|---|
| HTML | 게임판(canvas 또는 격자)과 점수 표시가 어디에 있는가 |
| CSS | 보드와 블록의 색, 화면 배치가 어디서 정해지는가 — 채팅에서는 보통 한 문서 안의 `<style>` 태그에 들어 있습니다 |
| JavaScript | 키 입력 처리, 블록 낙하, 줄 삭제가 어디에 있는가 — 보통 같은 문서의 `<script>` 태그에 들어 있습니다 |

한 가지 미리 알아둘 점: **같은 요청이라도 결과는 사람마다, 실행마다 다릅니다.** 블록 색, 조작감, 점수 규칙이 옆자리와 다를 수 있고, 코드 구성도 다를 수 있습니다. 어느 쪽이든 괜찮습니다. 어떤 결과든 그 안에 세 겹은 그대로 들어 있습니다. 동료들과 각자의 결과물을 공유하고 비교해 보세요. 동일한 "테트리스" 요청에 대해 AI가 조작 방식이나 디자인을 어떻게 서로 다르게 구현했는지 비교해 보는 것은 AI의 비결정적 특성과 프롬프트의 영향력을 이해하는 데 좋은 관찰이 됩니다.

> 🤔 **예측해보세요**: AI가 만든 첫 테트리스에 없는 것은 무엇일까요? 플레이하면서 "있으면 좋겠는데 없는 것"을 두 개 적어 보세요. 그것이 다음 단계의 재료입니다.

#### 부가 기능은 재요청으로 — 한 번에 하나씩

첫 결과물에는 보통 뼈대만 있습니다. 원하는 기능은 **한 번에 하나씩** 재요청으로 더합니다. 한꺼번에 다섯 개를 요청하면 무엇이 어디서 어긋났는지 추적하기 어렵습니다.

| 추가하고 싶은 것 | 재요청 예시 |
|---|---|
| 다음 블록 미리보기 | "다음에 나올 블록을 게임판 옆에 보여줘" |
| 점수와 레벨 | "줄을 지울 때마다 점수가 오르고, 10줄마다 낙하 속도가 빨라지게 해줘" |
| 일시정지 | "P 키를 누르면 일시정지되고, 화면에 '일시정지' 문구가 보이게 해줘" |
| 게임오버 후 다시하기 | "게임오버가 되면 최종 점수와 다시하기 버튼을 보여줘" |
| 색 테마 | "블록 색을 파스텔 톤으로 바꾸고 배경을 밝게 해줘" |

기능을 하나 더할 때마다 플레이해서 확인하고, **이전 기능이 망가지지 않았는지**도 함께 봅니다.

#### ✅ 1차 체크

| 확인 | ☐ |
|---|---|
| 채팅 미리보기에서 플레이된다 | ☐ |
| 세 겹이 어디에 있는지 위 표를 채웠다 | ☐ |
| 부가 기능을 최소 1개, 재요청으로 추가했다 | ☐ |
| 추가한 뒤에도 기존 플레이가 망가지지 않는 것을 확인했다 | ☐ |

마지막으로 하나 기억해 둡니다. 이 게임은 **대화 창 안에** 있습니다. 내 컴퓨터에는 파일이 하나도 생기지 않았습니다. 대화를 잃어버리면 게임도 함께 사라집니다. 이 차이가 2차의 출발점입니다.

---

2차는 방향이 반대입니다. 테트리스는 이름이 규칙 명세서였지만, 이번에 만들 것은 **내가 고른 게임** — 이름만으로는 AI에게 통하지 않는 게임입니다. 그래서 이번에는 규칙을 여러분이 적습니다. 도구도 바꿉니다. 1차의 게임은 채팅 창 안에 있었지만, 2차는 Day 4에 차린 작업대에서 **개발 에이전트(Claude Code 등)**로 만듭니다. 파일이 내 컴퓨터에 남아야 내일도, 다음 주에도 이어서 고칠 수 있습니다. 완성도 높은 게임이 아니라, 내가 설계한 규칙이 실제로 도는 것을 보는 산출물이 목표입니다.

게임을 만드는 이유가 있습니다. 소개 페이지는 만들고 나면 다시 볼 일이 드물지만, 게임은 만든 사람이 곧 첫 플레이어입니다. 누르고, 지고, 다시 하는 동안 세 겹이 전부 움직입니다. 그리고 규칙을 여러분이 직접 정하기 때문에, **AI가 규칙을 그럴듯하게 바꿔치기했는지 판정할 기준이 여러분 안에 있습니다.**

### 📐 게임이 되기 위한 최소 재료

장르가 무엇이든, 화면 위에서 돌아가는 게임에는 세 가지 재료가 있습니다.

- **보이는 것** — 점수판, 게임판, 버튼처럼 플레이어의 눈앞에 놓이는 요소들입니다.
- **상태** — 지금 몇 점인지, 진행 중인지 끝났는지, 이겼는지 졌는지 같은 "지금 상황"입니다.
- **규칙** — 플레이어가 무언가를 했을 때 상태가 어떻게 바뀌는지 정하는 약속입니다.

게임을 이 세 가지로 나눠 적어 두면, AI에게 무엇을 만들지 또렷이 전할 수 있고, 결과가 어긋났을 때 어느 부분이 문제인지 짚기도 수월합니다.

익숙한 게임 두 개를 분해해 보면 이렇습니다.

| | 보이는 것 | 상태 | 규칙 |
|---|---|---|---|
| 가위바위보 | 점수판, 가위·바위·보 버튼, 결과 문구 | 이김/짐/무승부, 선택한 버튼 강조 | 내 선택과 컴퓨터의 무작위 선택을 판정, 3승이면 종료 |
| 두더지 잡기 | 3×3 칸, 점수판, 남은 시간, 시작 버튼 | 두더지가 나온 칸, 잡았을 때 표시 | 무작위 칸에 두더지 등장, 클릭하면 +1점, 30초 뒤 종료 |

**한 화면 크기인지 가늠하는 기준**도 이 표로 잡을 수 있습니다. 각 칸이 한두 줄로 적히면 오늘 만들기에 적당한 크기입니다. 어느 칸이 문단이 되어 버리면 — 레벨, 저장, 랭킹까지 들어가기 시작하면 — 오늘 분량보다 큽니다. 규칙을 잘라 첫 판만 만들고, 나머지는 다음 루프에서 더하면 됩니다.

### 🎮 어떤 게임이든 좋습니다

한 화면에서 끝나고, 클릭이나 키보드로 조작하는 작은 게임이면 충분합니다. 고르기 어렵다면 아래에서 골라도 되고, 목록에 없는 게임이면 더 좋습니다.

- 가위바위보 (컴퓨터와 5판 승부)
- 숫자 맞히기 (업/다운 힌트)
- 반응속도 테스트 (신호가 바뀌면 최대한 빨리 클릭)
- 두더지 잡기 (제한시간 안에 나타나는 칸 클릭)
- 카드 뒤집기 기억력 게임 (같은 짝 찾기)

### 📝 먼저 내 게임을 세 가지로 나눠 보기

프롬프트를 쓰기 전에, 아래 표를 자기 게임으로 채웁니다. 이 표가 곧 요청문의 재료가 됩니다. 표를 채우기 어렵다면 게임이 아직 한 화면 크기보다 큰 것이므로, 규칙을 더 잘라냅니다.

| 재료 | 질문 | 내 게임에서는 |
|---|---|---|
| 보이는 것 | 화면에 무엇이 있어야 하나요? (점수판, 게임판, 버튼…) | ____ |
| 상태 | 어떤 상황이 눈에 보여야 하나요? (이김/짐, 진행 중, 선택됨…) | ____ |
| 규칙 | 무엇을 하면 무엇이 바뀌나요? 언제 끝나나요? | ____ |

### 🔧 2차 요청문 쓰기

위 표의 내용으로 대괄호를 채웁니다. 규칙은 여러분이 정합니다 — 테트리스와 달리 이 게임의 규칙은 AI 안에 없으므로, AI에게 규칙까지 맡기면 결과가 이상해도 무엇이 잘못인지 판정할 기준이 없습니다.

```text
새 폴더 my-game을 만들고, 그 안에 아래 게임을 만들어줘.

만들 게임: [예: 가위바위보 / 숫자 맞히기 / 내가 정한 게임]
규칙: [게임이 어떻게 진행되는지 2~3문장으로 직접 적기]
끝나는 조건: [예: 5판 승부 / 기회 10번 소진 / 제한시간 30초]

파일은 index.html, styles.css, script.js 세 개로 나눠줘.
- 화면: [내 게임의 화면 요소]를 담아줘.
- 모바일에서도 조작하기 편한 크기로, [내 게임의 상태]가
  색과 문구로 함께 보이게 해줘.
- 내가 적은 규칙 그대로 동작하고, 끝나는 조건이 되면
  결과와 다시하기 버튼이 나오게 해줘.

규칙에 애매한 부분이 있으면 만들기 전에 나에게 물어봐.
완료 후 내가 브라우저에서 확인할 방법과 수정 체크리스트를 알려줘.
```

중요한 점은 "세 파일로 나눠줘"가 아니라, 화면·상태·규칙을 구체적으로 적었다는 것입니다. 아래는 가위바위보를 예로 든 AI 출력 예시입니다. **여러분의 게임에서는 각 자리에 무엇이 오는지 대응시키며** 읽으세요.

### 📝 산출물 기록

실습이 끝나면 아래 표를 자기 노트에 채웁니다. 실패한 칸이 있어도 괜찮습니다. 실패 지점이 다음 지시의 재료입니다.

### 브라우저에서 플레이하며 확인하기

1. `index.html` 파일을 브라우저에서 열어 실행을 확인합니다. (파일을 마우스로 더블클릭하거나 브라우저 창으로 끌어다 놓으면 바로 열립니다.)
2. 이기는 경우와 지는 경우를 일부러 모두 만들어, 점수·문구·색이 함께 바뀌는지 확인합니다.
3. 버튼을 빠르게 연타해 점수가 이중으로 오르거나 화면이 꼬이지 않는지 확인합니다.
4. 브라우저 폭을 모바일 크기까지 줄여 버튼이 누르기 편한지 확인합니다.
5. 끝나는 조건까지 플레이해 결과와 다시하기가 나오는지, 다시하기 뒤 점수가 처음으로 돌아가는지 확인합니다.
6. `Tab` 키로 버튼 사이를 이동해 키보드만으로도 플레이할 수 있는지, 개발자도구 Console에 오류가 없는지 확인합니다.

## 바이브 코딩 루프: 계획하고, 만들고, 다듬기

---

AI와 일하는 리듬은 한 방향이 아니라 루프입니다. **계획을 맞추고 → 만들게 하고 → 결과를 관찰하고 → 다시 요청하고**, 이 네 걸음을 돕니다. 한 번에 완벽한 결과를 노리기보다 이 루프를 빠르게 여러 번 도는 것이 바이브 코딩의 힘입니다. 앞에서 게임을 한 번 만들어 봤으니, 이번에는 그 앞뒤에 붙는 두 요령 — **시작 전 계획 맞추기**와 **결과 보고 다시 요청하기** — 를 익힙니다.

### 실제 개발자처럼: 만들기 전에 PRD부터 쓰기

전문 개발팀은 코드를 짜기 전에 **PRD(Product Requirements Document, 제품 요구사항 문서)** 를 씁니다. "무엇을, 누구를 위해, 어떤 규칙으로 만드는가"를 먼저 글로 못 박아 두는 문서입니다. 머릿속으로만 그리고 바로 만들면 사람마다 다른 그림을 그리게 되니, 만들기 전에 한 장짜리 합의를 만들어 두는 것입니다.

바이브 코딩에서는 이 습관이 오히려 더 중요합니다. AI는 내 **마지막 한 마디**에 맞춰 움직이기 때문에, 목표가 글로 고정돼 있지 않으면 대화가 길어질수록 방향이 조금씩 흘러갑니다. 그래서 요즘은 "먼저 명세(PRD)를 쓰고, 그다음 AI에게 시킨다"는 방식(스펙 주도 개발, spec-driven development)이 널리 권장됩니다.

거창할 필요는 없습니다. 만들기 전 15~30분, 한 장이면 충분합니다. 앞 섹션에서 게임을 화면·상태·규칙으로 나눠 적어 본 표가 사실 이 PRD의 뼈대입니다. 여기에 몇 가지만 더합니다.

- **무엇을**: 어떤 게임인지 한 줄 (예: 두더지 잡기)
- **누가**: 누가 플레이하나
- **규칙과 끝나는 조건**: 어떻게 진행되고 언제 끝나는가
- **화면 요소**: 무엇이 보여야 하나 (점수판·게임판·버튼…)
- **예외 상황**: 어긋났을 때 어떻게 (시간이 다 됐는데 클릭하면? 같은 칸을 두 번 누르면?)

PRD 초안도 AI와 함께 잡아도 됩니다. "이런 게임을 만들 건데 PRD 초안을 잡아 줘"라고 한 뒤, 틀린 가정만 내가 바로잡으면 됩니다. 그리고 곧바로 코드를 시키지 말고 계획부터 확인합니다.

```text
바로 코드를 쓰지 말고, 위 PRD대로 어떻게 만들지 계획을 먼저 알려줘.
- 어떤 순서로 만들지
- 내가 미리 정해줘야 할 것이 있는지
계획이 내 생각과 맞으면 "진행"이라고 할게. 그때 만들어줘.
```

이렇게 하면 AI가 내 요청을 어떻게 이해했는지 **코드가 나오기 전에** 확인할 수 있고, 방향이 틀렸을 때 문서 한 줄만 고치면 되지 수백 줄 코드를 되돌릴 필요가 없습니다.

> 💡 손으로 코드를 빨리 쓰는 힘은 이미 AI가 가졌습니다. 남는 실력은 **무엇을 만들지 또렷이 정의하는 힘** — 좋은 명세를 쓰는 사람이 좋은 결과를 얻습니다.

### 🔍 결과를 관찰하고 다시 요청하기

계획대로 만들어도 첫 결과가 한 번에 마음에 꼭 들지는 않습니다. 그게 정상이고, 여기서 다듬는 루프가 시작됩니다. 이때 핵심은 문제를 정확히 관찰해서, 무엇이 어떻게 되면 좋겠는지 말하는 것입니다.

재요청은 "어떤 CSS 속성을 써라" 같은 해법을 외워서 말하는 것이 아닙니다. 그건 AI의 일입니다. 여러분은 **"어디서 무슨 문제를 봤고, 어떻게 되길 원한다"** 를 평소 쓰는 말로 전하면 충분합니다. 그 말을 듣고 어느 파일을 어떻게 고칠지는 AI가 알아서 정합니다.

재요청의 기본 틀은 한 문장입니다.

> **[어디서] [무슨 문제가 있고], [어떻게 되면 좋겠다].**

아래는 그대로 가져다 써도 되는 재요청 문장의 예입니다. 문제(관찰)와 바라는 결과(기대)만 담고, 해법은 넣지 않았습니다.

- "결과 문구가 무슨 뜻인지 알기 어려워. '이겼습니다/졌습니다'처럼 분명하게 바꿔줘."
- "모바일에서 버튼이 좁아져서 누르기 불편해. 버튼이 커지거나 줄바꿈되어 누르기 편하게 해줘."
- "무승부인데도 내 점수가 올라가. 이겼을 때만 점수가 오르게 해줘."

AI가 고쳐서 돌려주면, 내가 말한 그 부분이 실제로 바뀌었는지 직접 플레이하거나 코드를 열어 확인하세요. 무작정 믿지 않는 이 습관이 바이브 코딩의 핵심입니다.

"전체를 다시 만들어줘"보다 문제가 있는 부분만 짚는 편이 낫습니다. 범위가 좁으면 이미 괜찮은 부분이 덜 흔들립니다. 해법까지 지정할 필요는 없습니다 — "모바일 간격" 정도로 말해도 AI가 알아듣습니다.

### 📝 판단 워크북

| 화면은 괜찮지만 키보드로 버튼을 이동하기 어렵습니다 | ____ | ____ |

<details>
<summary>💡 검토 포인트</summary>

- 규칙이 전달되지 않는 것은 화면의 글과 정보 문제입니다. 안내 문구와 구조를 손보게 요청합니다.
- 작은 화면에서 글자가 잘리는 문제는 크기와 배치, 줄바꿈을 손보게 요청합니다.
- 게임이 끝나지 않는 문제는 종료 조건 처리를 손보게 요청하고, 결과·다시하기 화면이 실제로 나타나는지 함께 확인합니다.
- 키보드 이동은 화면 구조와 동작, 표시가 함께 걸리는 문제입니다. 이 경우 한 곳만이 아니라 관련된 부분을 함께 봐 달라고 요청합니다.

</details>

> **[과제] 과제 제출 — 나만의 게임을 웹에 올리고 링크 제출하기**
>
> - **무엇을** 오늘 만든 게임을 GitHub Pages로 배포해 누구나 링크만으로 열 수 있게 만듭니다.
> - **무엇으로** 실제 플레이가 되는 웹 URL 하나를 제출합니다 (예: https://내아이디.github.io/my-game/).
> - **인정** 새 창·시크릿 창에서 게임이 뜨고, 그 주소에서 조작이 실제로 동작하는 것입니다.
> - **예** 배포 직후 주소가 뜨기까지 1~2분 걸릴 수 있습니다. 휴대폰으로도 열리는지, 동료에게 링크를 보내도 열리는지 확인해 보세요.
>
> 마감: 2026-11-03T23:59+09:00
>
> **평가 기준**
>
> | 기준 | 이렇게 하면 충족 | 배점 |
> | --- | --- | --- |
> | 배포 완료 | 제출한 URL 을 새 창에서 열면 게임이 뜬다. | 40 |
> | 동작 확인 | 내 컴퓨터가 아니라 그 웹 주소에서 조작이 실제로 동작한다. | 30 |
> | 접근 확인 | 휴대폰 또는 동료가 보낸 링크를 열어 게임을 플레이할 수 있다. | 30 |
>
> 제출은 LMS 레슨 화면의 과제 카드에서 GitHub 또는 Google Drive 링크로 합니다(이 파일에 적으면 제출되지 않습니다).

오늘 만든 나만의 게임은 아직 내 컴퓨터(또는 대화창) 안에만 있습니다. 이번 과제에서는 그 게임을 웹에 올려 **누구나 링크만으로 열어볼 수 있게** 만들고, 그 링크를 제출합니다. '나만 보는 결과물'이 '세상에 내보내는 서비스'로 바뀌는 첫걸음입니다.

### 🎮 무엇을 제출하나요

- 오늘 만든 나만의 게임(가위바위보, 숫자 맞히기, 두더지 잡기 등 무엇이든 좋습니다). 시간이 남으면 기능을 더 다듬거나 추가해도 됩니다.
- 그 게임이 실제로 열리는 **웹 링크 하나**.

### 🚀 GitHub Pages로 배포하기

배포도 바이브 코딩으로 합니다. **GitHub Pages**는 HTML·CSS·JS로 만든 페이지를 무료로 웹에 올려 주는 서비스입니다. 파일을 GitHub에 올리고 Pages를 켜면 `https://내아이디.github.io/저장소이름/` 같은 주소가 생기고, 그 주소로 전 세계 누구나 접속할 수 있습니다.

**준비물**: GitHub 계정 하나. 없다면 [github.com](https://github.com)에서 무료로 가입합니다.

작업대(Day 4에서 차린 Claude Code)에서 게임 폴더를 열고 이렇게 요청합니다.

```text
이 폴더의 게임을 GitHub Pages로 배포하고 싶어.

1. 이 폴더를 깃 저장소로 만들고, 내 GitHub에 새 공개 저장소로 올려줘.
2. GitHub Pages를 켜서 웹에서 열 수 있는 주소를 만들어줘.
3. 로그인이나 권한 확인처럼 내가 직접 눌러야 하는 단계는 멈추고 알려줘.
4. 배포가 끝나면 접속 주소를 알려주고, 잘 열리는지 확인하는 방법도 알려줘.
```

> ⚠️ 배포에는 계정 로그인·권한 승인처럼 **사람이 직접 해야 하는 단계**가 섞여 있습니다. AI가 그 대목에서 멈추고 안내하면, 그 부분만 웹 화면에서 처리하고 다시 이어가면 됩니다. 배포 직후에는 주소가 뜨기까지 1~2분 걸릴 수 있습니다.

### ✅ 제출 전 체크

| 확인 | ☐ |
|---|---|
| 링크를 새 창(또는 시크릿 창)에서 열었을 때 게임이 뜬다 | ☐ |
| 내 컴퓨터가 아니라 그 웹 주소에서 조작이 실제로 동작한다 | ☐ |
| 휴대폰으로 열어도 플레이할 수 있다 | ☐ |
| 링크를 동료에게 보냈을 때도 열린다 | ☐ |

### 📮 제출

- **제출물**: 배포된 게임 링크 (예: `https://내아이디.github.io/my-game/`)
- 제출 방법은 과정 안내(LMS 과제 제출란 또는 지정 채널)를 따릅니다.

### 🤔 마무리 3문장

2. 배포 과정에서 AI가 멈추고 나에게 넘긴 단계는 무엇이었나요? 왜 그 단계는 사람이 해야 했을까요?
3. 이 게임에 딱 한 가지만 더한다면 무엇을 더하고 싶나요?
