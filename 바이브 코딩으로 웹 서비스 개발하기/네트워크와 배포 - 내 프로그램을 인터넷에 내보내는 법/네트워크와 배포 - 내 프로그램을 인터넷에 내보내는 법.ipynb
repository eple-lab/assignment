### 여는 이야기 - 친구에게 내 서비스 링크를 보냈는데

할 일을 기록하는 Todo 앱은 내 컴퓨터에서만 열립니다. 데이터를 저장하고 다시 불러오는 것까지 확인했으니, 이제 친구에게 보여 주고 싶습니다.

> **나**: 이거 봐! 내가 만든 Todo 앱이야. `http://localhost:3000`
> **친구**: 안 열리는데? "이 사이트에 연결할 수 없음"이라고 떠.
> **나**: 내 컴퓨터에서는 잘 열리는데, 왜 그럴까?

데이터를 남기는 일과 다른 사람이 접속하게 하는 일은 다릅니다. 오늘은 내 컴퓨터 안에서 돌던 서비스를 인터넷으로 내보내 봅니다.

### 오늘의 핵심 한 문장

> **배포는 내 컴퓨터에서 돌던 서비스를 다른 사람도 접속할 수 있는 주소에서 실행하는 일입니다.**

오늘의 결과물은 **배포된 내 서비스 URL** 입니다. 친구의 브라우저에서도 서비스가 열리는지, 화면에서 보낸 요청에 백엔드가 응답하는지 확인하고 그 URL을 기록합니다.

## 주소를 넣으면 화면이 뜨기까지 - 인터넷의 뼈대

---

![전 세계 바닷속에 깔린 해저 광케이블 지도](https://commons.wikimedia.org/wiki/Special:FilePath/Submarine_cable_map_umap.png?width=640)
*인터넷은 구름이 아닙니다 - 대륙과 대륙을 잇는 이 해저 케이블들을 타고 데이터가 실제로 흐릅니다. (이미지: Wikimedia Commons)*

백엔드에 요청을 보내려면 어디로 보낼지 알고, 응답을 읽고, 안전하게 연결됐는지 확인해야 해요. 네트워크 전체를 공부하는 대신, 배포할 때 필요한 질문을 세 묶음으로 나눠 볼게요.

- __① 주소와 도달성__: IP·도메인·DNS로 목적지를 찾고, localhost·사설 IP·NAT·포트로 그곳까지 접속할 수 있는지 살펴봐요.
- __② 요청과 응답__: HTTP로 무엇을 요청하는지, 상태 코드로 어떤 결과를 받았는지 읽어요.
- __③ 안전한 통신__: HTTPS와 인증서로 연결 상대와 통신 보호를 확인해요.

`mysite.com`을 열었을 때 화면이 보이지 않는다면, 어느 단계에서 멈췄는지 구분하는 것이 목표예요. 먼저 주소부터 볼게요.

### ① 주소와 도달성

#### IP·도메인·DNS: 어디로 보낼까요?

인터넷의 모든 컴퓨터에는 __번호(IP 주소)__ 가 있습니다. 예를 들면 `142.250.72.14` 같은 숫자죠. 문제는, 이 숫자를 일일이 외우고 관리하기가 여간 불편한 게 아니라는 것입니다 - IPv6 주소는 `2400:cb00:2048:1::c629:d7a2`처럼 더 길고, 서버 사정에 따라 수시로 바뀌기도 합니다.

- __도메인__(`mysite.com`) = 사람이 외우는 __가게 이름__. 전화번호 대신 "김밥천국"이라고 부르는 것과 같습니다.
- __DNS(Domain Name System)__ = 이름을 번호로 바꿔 주는 __전화번호부__. 브라우저가 "`mysite.com`이 어디예요?" 하고 물으면, DNS가 "`142.250.x.x`입니다" 하고 IP를 알려 줍니다.

__도메인은 사는 것__ 입니다. 1년에 만 원 안팎이면 `내이름.com` 같은 주소를 등록할 수 있습니다. 등록한 뒤에는 "이 도메인은 저 서버(IP)를 가리켜라"라고 __연결(DNS 설정)__ 해 줍니다.

> __흔한 함정 - 왜 바로 안 바뀌지?__ 도메인이 가리키는 곳을 바꿔도 __즉시 반영되지 않을 때__ 가 있습니다. DNS는 답을 한동안 기억(캐시)해 두기 때문에, 어떤 사람은 새 주소를, 어떤 사람은 옛 주소를 봅니다. 보통 몇 분에서 몇 시간이면 전 세계로 퍼집니다(이걸 '전파'라고 합니다). "방금 바꿨는데 왜 안 바뀌지?"의 가장 흔한 정답입니다. (출처: DCHost, "What Is Web Hosting")

#### 도메인은 '별칭' 그 이상 - 왜 이름 한 겹을 두는가

도메인을 '외우기 편한 IP 별명'으로만 알면 절반만 아는 것입니다. 도메인의 진짜 힘은 __이름과 실제 서버 사이에 '한 겹'을 두는 것__ - 바로 이 과정에서 거듭 만난 추상화입니다. 앞(도메인)은 그대로 두고 뒤(서버·IP)를 얼마든지 바꿀 수 있죠. 다음은 이름을 따로 두었을 때 얻는 이점이에요. (출처: Cloudflare, "What is DNS?")

- __① 서버를 옮겨도 주소는 그대로__ - 호스팅을 바꾸거나 서버가 죽어 새 IP를 받아도, 고치는 건 'DNS 한 줄'뿐이고 세상에 뿌린 주소는 손대지 않습니다. 가게가 이사를 가도 상호는 그대로라, 단골은 상호만 기억하면 되는 것과 같죠.
- __② 하나의 이름 → 여러 서버__ - 한 도메인에 서버를 여러 대 물려, 접속을 나눠 보내고(부하 분산) 지역별로 가까운 서버로 안내합니다(CDN). 큰 서비스가 트래픽이 몰려도, 서버 하나가 죽어도 잘 안 멈추는 비결의 일부입니다. (출처: Cloudflare, "DNS load balancing")
- __③ 서브도메인으로 구조화__ - 하나의 도메인 아래 `api.`·`blog.`·`shop.`처럼 여러 서비스를 갈라 둘 수 있습니다. 각자 다른 서버로 보내도 브랜드는 하나로 묶이죠.
- __④ 내 이름으로 보유·이동__ - 도메인은 등록해 내 이름으로 갖고(정확히는 갱신하며 쓰는 '임대'에 가깝습니다), 호스팅 업체를 바꿔도 그대로 들고 갑니다. 그래서 브랜드 자산이 됩니다. (출처: ICANN)

#### localhost·사설 IP·NAT: 내 컴퓨터까지 닿을까요?

처음에 친구가 열지 못했던 `localhost:3000`부터 다시 볼게요. __친구의 `localhost`는 친구 컴퓨터를 가리켜요.__ 내 서버를 찾는 주소가 아니므로, 친구에게 같은 링크를 보내도 내 서버로 연결되지 않아요. 이건 공유기가 막아서 생긴 문제가 아니라 __주소가 가리키는 대상이 다른 문제__ 예요.

그렇다면 내 컴퓨터의 다른 주소를 알려 주면 될까요? 여기서는 내부망과 외부망의 구분도 알아야 해요.

- __내부망(LAN)__ - 집이나 회사 안에서 공유기 등으로 묶인 네트워크예요. 가정에서는 보통 `192.168.0.5` 같은 __사설 IP__ 를 써요. 같은 내부망에서 기기끼리 통신할 때 쓰는 주소이지, 인터넷 어디서나 내 컴퓨터를 찾을 수 있는 주소는 아니에요.
- __외부망(WAN)__ - 우리 내부망 바깥의 넓은 네트워크예요. 집 공유기를 통해 인터넷으로 나간다고 생각하면 돼요. 일반적인 가정용 구성에서는 여러 기기가 공유기의 __공인 IP 하나__ 를 함께 써요.

공유기는 __NAT__ 라는 기능으로 내부 기기의 사설 주소와 외부 통신에 쓰는 주소를 바꿔 연결해 줘요. 내부에서 시작한 요청의 응답은 돌아올 수 있지만, 외부에서 갑자기 들어온 요청을 내 노트북의 어느 프로그램으로 보낼지는 별도 설정 없이는 정해져 있지 않아요. 방화벽 설정도 접속에 영향을 줘요. 그러니 __외부에서 내 컴퓨터까지 들어오는 길이 있나__ 는 `localhost`와 구별해서 확인할 문제예요.

> 친구도 쓸 수 있는 서비스를 만들려면 __친구가 접속할 주소와 실제 서버까지 이어지는 길__ 이 필요해요. 클라우드 서버에 배포해 공개 URL을 받거나, 내 컴퓨터의 서버를 터널로 바깥과 연결하는 방법이 있어요. 뒤의 '집 서버' 섹션에서 터널을 살펴볼 거예요.

#### 포트(PORT)와 방화벽: 실행 중인 프로그램까지 닿을까요?

한 대의 서버 컴퓨터는 여러 가지 일을 동시에 합니다(웹·메일·데이터베이스…). 그래서 "이 요청은 어느 창구로 들어와라"를 정하는데, 그 창구 번호가 __포트__ 입니다.

- __IP가 '건물 주소'라면, 포트는 그 건물 안의 '호실 번호'__ 입니다. 컴퓨터 한 대(IP 하나)에 문이 수만 개 있고(0～65535번), 프로그램마다 자기 문에서 손님을 받습니다.
- 웹은 보통 __80번(HTTP)·443번(HTTPS)__ 문을 씁니다. 개발 중 자주 보는 `localhost:3000`에서 __3000이 바로 포트 번호__ 예요 - "이 컴퓨터의 3000번 문"이라는 뜻이죠.
- __방화벽__ 은 그 문을 지키는 경비입니다. 필요 없는 문은 닫아 두어 아무나 못 들어오게 막습니다.

그래서 "내 컴퓨터에선 되는데 밖에선 안 열려요"의 흔한 원인이 이겁니다 - 열려야 할 문(포트)이 방화벽에 막혔거나 안 열린 것. 배포 플랫폼을 쓰면 대개 알아서 열어 주니 직접 만질 일은 드뭅니다. 다만 "밖에서 안 열린다"면 __이 문이 닫혔을 수 있다__ 는 개념만 기억해 두세요. (출처: WeWeb, "Self-Hosting a Web App in 2026")

### ② 요청과 응답

#### HTTP: 무엇을 요청하고 돌려받을까요?

브라우저와 서버가 대화하려면 __공통 언어__ 가 필요합니다. 그게 __HTTP(HyperText Transfer Protocol)__ 입니다.

- __HTTP__ = "이 페이지 줘"(요청) → "여기 있어"(응답)로 오가는 규칙. 웹의 탄생과 함께 만들어진 언어입니다.
- 실제 요청/응답은 이런 모습에 가깝습니다(개념만 보세요).

```text
[브라우저의 요청]
GET  /products      ← "products(상품) 목록 페이지를 줘(GET)"
Host: mysite.com

[서버의 응답]
200 OK              ← "성공! 여기 있어"
(그리고 화면 내용이나 데이터가 따라옵니다)
```

HTTP 대화가 도착했다면 이제 서버의 응답을 읽을 차례예요. 응답 첫머리의 숫자부터 살펴볼게요.

#### 상태 코드: 응답 결과를 읽는 법

서버는 응답할 때 __세 자리 숫자__ 를 함께 보냅니다. 이게 __상태 코드__ 입니다. 방금 응답 예시의 `200 OK`가 바로 그것이죠. 수백 개를 외울 필요 없이, __맨 앞 숫자__ 만 보면 됩니다.

- __2xx = 성공__ 초록불, 통과
- __3xx = 다른 데로__ 노란불, 우회(주소가 옮겨졌을 때)
- __4xx = 요청 쪽 문제일 가능성이 큼__ 주소가 틀렸거나 권한이 없는 경우
  예를 들어 403은 서버의 권한 설정 문제로도 생길 수 있으니, 요청한 사람의 잘못으로 단정하지 마세요.
- __5xx = 서버의 잘못__ 서버가 터짐

가장 자주 만나는 스타 플레이어들:

| 코드 | 뜻 | 한마디로 |
|---|---|---|
| __200 OK__ | 성공 | "다 잘됐어" (잘 되면 눈에 안 보이는 코드) |
| __301/302__ | 주소 이동 | "그 페이지 저쪽으로 옮겼어" (자동으로 새 주소로 보냄) |
| __401 Unauthorized__ | 로그인 필요 | "__누구세요?__" (아직 로그인 안 함) |
| __403 Forbidden__ | 권한 없음 | "누군지 알지만, __넌 안 돼__" (VIP 바운서) |
| __404 Not Found__ | 없는 페이지 | "그런 페이지 __없어__" (인터넷에서 제일 유명한 에러) |
| __429 Too Many Requests__ | 너무 많은 요청 | "__천천히 좀__" (짧은 시간에 너무 자주 요청) |
| __500 Internal Server Error__ | 서버 오류 | "__네 잘못 아냐, 내가 터졌어__" |

> __재미 하나__ - __418 "I'm a teapot"__(나는 찻주전자)이라는 코드도 있습니다. 1998년 만우절 농담으로 만든 것으로, "나는 찻주전자라서 커피를 못 내려"라는 뜻입니다. 실제 서버에 쓰라고 만든 건 아니고, 구글에도 이스터에그로 숨어 있어요(`google.com/teapot`). 참고로 __451__ 이라는 코드는 소설 『화씨 451』에서 따와 "법적 이유로 막힌 콘텐츠"를 뜻합니다. 엔지니어들의 유머 감각이죠. (출처: MDN Web Docs, "HTTP response status codes")

> __실전 팁__ - __4xx면 내 요청을 고치고, 5xx면 서버(만든 사람) 문제__ 입니다. AI에게 "500 에러가 나" 하고 화면의 에러를 그대로 붙여 주면 원인을 훨씬 잘 찾아 줍니다.

답은 마음속으로 정해 보세요.

- 주소를 잘못 입력해 없는 페이지를 요청했다
- 로그인도 안 했는데 내 정보 페이지에 들어가려 했다
- 로그인은 했지만 남의 글을 __수정__ 하려 했다
- 서버 코드에 버그가 있어 처리 중 터졌다
- 페이지가 정상적으로 잘 열렸다

| 주소를 잘못 입력해 없는 페이지를 요청했다 | 404 |
| 로그인도 안 했는데 내 정보 페이지에 들어가려 했다 | 401("누구세요?") |
| 로그인은 했지만 남의 글을 __수정__ 하려 했다 | 403("넌 안 돼") |
| 서버 코드에 버그가 있어 처리 중 터졌다 | 500 |
| 페이지가 정상적으로 잘 열렸다 | 200 |

이 섹션 ② '요청과 응답'에서 본 상태 코드 분류를 기준으로 답을 비교해 보세요. 401과 403의 차이(누구세요? vs 넌 안 돼)는 다음 섹션의 '인증 vs 인가'에서 다시 만납니다.

</details>

#### 직접 해보기: 상태 코드를 눈으로 확인하기

방금 배운 상태 코드, 지금 바로 볼 수 있습니다. 아무 웹사이트에서:

1. 화면에서 __마우스 오른쪽 클릭 → '검사'(Inspect)__ 를 누릅니다.
2. 위쪽 탭에서 __'네트워크(Network)'__ 를 고르고, 페이지를 __새로고침__ 합니다.
3. 주르륵 뜨는 목록의 __맨 위 줄__ 을 보세요 - `Status` 칸에 `200`(성공) 같은 숫자가 보입니다.

브라우저와 서버가 방금 나눈 대화가 여기 다 기록돼 있습니다. 화면이 안 뜰 때 이 탭을 열어 __빨간 줄(4xx·5xx)__ 을 찾으면 어디가 문제인지 단서가 나옵니다. AI에게 도움을 청할 때 이 정보를 함께 주면 훨씬 정확한 답을 받습니다. (출처: MDN Web Docs, "HTTP response status codes")

### ③ 안전한 통신

#### HTTPS와 인증서: 누구와 안전하게 대화하나요?

이 섹션 ①에서 본 주소 찾기와 ②에서 본 HTTP 대화만으로는 통신 내용이 보호되지 않아요. __HTTPS__ 는 HTTP 대화를 암호화하고 전송 중 변조를 확인하며 보호하는 방식이에요. 이때 사용하는 보안 기술을 __TLS__ 라고 불러요. 엽서를 그대로 보내는 대신 내용을 잠근 봉투에 담는다고 생각하면 돼요.

그런데 대화를 잠그기 전에, 상대가 내가 입력한 도메인에 맞는 서버인지도 확인해야겠죠? 이때 서버가 보여주는 것이 __인증서__ 예요. 인증서에는 적용되는 도메인 이름, 공개키, 유효기간 등이 담겨 있고, __인증기관(CA)__ 의 서명이 붙어 있어요. 인증기관은 도메인을 통제할 권한 등을 확인하고 인증서를 발급해요. (출처: Cloudflare, "Domain Control Validation")

브라우저가 `https://mysite.com`에 연결할 때는 이렇게 확인해요.

1. 서버가 제시한 인증서에 __접속한 도메인 이름이 맞는지__ 확인해요.
2. 인증서가 __유효기간 안에 있고 신뢰할 수 있는 인증기관의 서명으로 이어지는지__ 확인해요.
3. 연결 과정에서 서버가 인증서의 공개키에 대응하는 __개인키를 가지고 있음을 확인__ 하고, 통신에 쓸 암호 키를 마련해요.
4. 이후 요청과 응답을 암호화해서 보내고, 전송 중 내용이 바뀌지 않았는지도 확인해요.

이 연결 준비 과정을 __TLS 핸드셰이크__ 라고 해요. 인증서는 상대 도메인을 확인하는 데 쓰고, 암호화는 대화 내용을 보호하는 데 쓰는 거예요.

> __HTTPS는 '믿을 만한 회사'라는 증명이 아니에요.__ 가짜 도메인을 가진 피싱 사이트도 자기 도메인에 유효한 인증서를 받을 수 있어요. 연결이 암호화돼 있다는 것과 그 사이트의 운영자를 믿어도 된다는 것은 다른 문제예요. 주소창의 연결 표시뿐 아니라 __도메인 철자 자체__ 도 꼭 확인하세요. 브라우저에 따라 자물쇠 대신 다른 아이콘으로 연결 정보를 보여주기도 해요. (출처: Cloudflare, "Domain spoofing")

> 많은 배포 플랫폼은 인증서 발급과 갱신을 자동으로 처리해 줘요. 여러분은 공개 URL이 __HTTPS로 열리는지__, 인증서 경고가 없는지, 의도한 도메인이 맞는지 확인하면 돼요. 인증서 경고가 뜨면 무시하고 진행하지 말고 도메인 연결과 인증서 설정을 확인하세요.

<details class="lms-extra">

<summary>선택 읽기: 공개키·개인키와 통신용 키의 역할</summary>

- __공개키__ 는 인증서와 함께 공개해도 돼요. 브라우저가 서버의 증명을 확인할 때 사용해요.
- __개인키__ 는 서버 쪽에서 비밀로 보관해요. 서버는 이것으로 연결 과정에 서명해서, 인증서에 대응하는 개인키를 가진 상대임을 증명해요.

인증기관의 서명은 __인증서를 확인하는 데__ 쓰고, 서버의 서명은 지금 연결하는 상대가 그 개인키를 가졌는지 확인하는 데 쓴다고 구별하면 돼요. 개인키 자체를 브라우저에 보내는 건 아니에요.

그리고 실제 요청과 응답을 매번 이 공개키로 잠그는 것은 아니에요. 현대적인 TLS 연결에서는 핸드셰이크 중에 __이번 연결에 사용할 암호 키를 양쪽이 안전하게 마련__ 하고, 이후 대화는 그 키를 쓰는 빠른 방식으로 보호해요. 공개키로 비밀 열쇠를 그냥 보내 주는 과정이라고 외우기보다, __상대를 확인하고 통신용 키를 마련한 뒤 대화한다__ 는 역할을 기억하세요. (출처: Cloudflare, "How does public key encryption work?")

</details>

### 배포 후에는 세 묶음으로 확인해요

친구에게 `https://mysite.com`을 보냈다고 생각해 보세요. 링크가 열리지 않으면 아래 순서로 관찰한 사실을 모아 보세요.

1. 이 섹션 ①에서 본 주소와 도달성을 기준으로, 도메인이 의도한 서버를 가리키고 443번 포트로 접속할 수 있는지 확인해요.
2. 이 섹션 ②에서 본 요청·응답과 상태 코드를 네트워크 탭에서 확인해요.
3. 이 섹션 ③에서 본 HTTPS·인증서 확인 항목을 기준으로, 브라우저의 연결 경고가 있는지 확인해요.

이것은 문제를 점검하는 순서이지, 통신이 일어나는 시간순서는 아니에요. HTTPS 연결 준비가 끝난 뒤에 HTTP 요청을 보내게 돼요. 실제 데이터가 필요할 때 브라우저의 요청을 받아 처리하고 응답하는 쪽이 백엔드예요. 서버는 요청에 맞게 데이터를 조회하거나 저장하고, 그 결과를 응답으로 돌려줘요.

### 작동 원리: 패킷 - 데이터는 잘게 쪼개져 여행한다

사진 한 장을 보낼 때, 그게 통째로 인터넷을 건너갈까요? 아닙니다. 데이터는 __작은 조각(패킷)으로 쪼개져__ 각자 여행한 뒤, 도착지에서 __다시 조립__ 됩니다. 세 가지 비유면 충분합니다.

- __우편 비유__ - 긴 편지를 인덱스카드 여러 장에 나눠 적고, 각 카드에 '보낸 사람·받는 사람·몇 번째 장'을 표시해 보냅니다. 받는 쪽이 번호 순서대로 맞춰 읽죠. 패킷도 똑같이 __머리표(주소·순번)__ 와 __내용__ 을 담습니다.
- __피자 비유__ - 피자를 통째로 보내다 떨어뜨리면 다 망하지만, 조각으로 나눠 보내면 한 조각만 없어져도 __그 조각만 다시__ 보내면 됩니다.
- __내비 비유__ - 패킷들은 __서로 다른 길__ 로 갑니다. 한 길이 막히면 다른 길로 우회하죠. 그래서 인터넷 일부가 고장 나도 전체가 멈추지 않습니다.

이 방식 덕분에 __수십억 대의 기기가 같은 회선을 동시에__ 나눠 쓸 수 있습니다. 데이터를 쪼개 주소를 붙이고 순서를 맞추는 이 공통 규칙을 __TCP/IP__ 라고 부릅니다 - 인터넷 역사에서 봤던 바로 그 약속입니다. (출처: Cloudflare, "What is a packet?")

## 프로토타입과 프로덕트 사이 - AI가 약한 곳

## 화면은 금방 완성되는데, 왜 거기서 막힐까 - AI 코드의 약점 3가지

---

바이브 코딩에는 아주 흔한 실패 패턴이 있습니다. 이런 식이죠.

> 첫날: "우와, 하루 만에 화면이 다 만들어졌어!"
> 다음 주: "로그인이 자꾸 풀려… 저장한 게 사라져… 어? 클라우드 요금이 왜 이렇게 나왔지?"

**화면은 금방 살아나지만, 그다음 벽에 부딪힙니다.** 왜일까요? AI는 **"돌아가는 코드"** 는 잘 만들지만 **"안전하고 튼튼한 코드"** 는 놓치기 때문입니다. AI는 여러분이 **명시적으로 요구하지 않으면** 보안을 알아서 챙기지 않습니다. (출처: Sashido, "Vibe Coding to Production: The Backend Reality Check")

그래서 이 세 가지는 **여러분이 반드시 챙겨야** 합니다. 문법을 외울 필요는 없고, AI에게 **요구** 할 줄만 알면 됩니다. 하나씩 봅시다.

### ① 보안 - AI 코드의 최대 약점

보안 회사 **Veracode** 가 100개 넘는 AI 모델로 실험했더니, **AI가 생성한 코드의 약 45%에 보안 결함** 이 있었습니다. 게다가 모델이 더 똑똑해져도 이 비율은 거의 줄지 않았습니다 - 문법은 좋아졌지만 **안전은 그대로** 라는 뜻입니다. (출처: Veracode, "2025 GenAI Code Security Report")

가장 흔한 구멍 하나를 장면으로 봅시다. **"입력을 그대로 믿는" 실수** 입니다.

> 검색창에 사용자가 이름을 칩니다. 보통은 "김민준"이라고 치죠. 그런데 나쁜 사람은 이름 칸에 **명령어처럼 생긴 문자열** 을 넣습니다. 서버가 그걸 그대로 믿고 실행 하면, 데이터베이스가 통째로 털리거나 지워질 수 있습니다.

그래서 규칙은 하나 - **"사용자가 보낸 값은 일단 의심하고 검사한다."** 프롬프트로 요구하는 것이 출발점이고, 결과가 지켜졌는지 점검하고 시험해 봐야 해요.

> "사용자 입력을 **검증** 하고, 비밀번호는 해시로 저장 하고, 모든 통신은 HTTPS 로 해줘. 흔한 보안 취약점(입력값 주입, XSS)을 막아줘."

이 요구가 가장 흔한 구멍을 줄여 주지만, 결과는 반드시 점검해야 해요.

### 보안 이야기: 역사에 남은 데이터 유출 사고들

'보안이 왜 중요한지'는 실제 사고를 보면 확 와닿습니다. 유명한 사례 몇 가지 - 대부분 **막을 수 있었던** 것들입니다.

- **에퀴팩스(2017)** - 미국 신용평가사. 알려진 취약점을 방치하다 **약 1억 4,300만 명** 의 주민번호·생일·주소가 털렸습니다. 이미 공개돼 경고된 취약점을 방치한 게 화근이었습니다.
- **소니 플레이스테이션 네트워크(2011)** - **약 7,700만 명** 의 정보가 유출돼 서비스가 오래 멈췄습니다.
- **야후 보이스(2012)** - 비밀번호를 **암호화 없이 그대로(평문)** 저장했다가 통째로 유출됐습니다. (앞서 "비밀번호는 해시로"를 강조한 이유가 여기 있습니다.)
- **마이스페이스(2008)** - 약 3억 6천만 계정 유출. 교훈: **여러 사이트에 같은 비밀번호를 쓰지 말 것.** 한 곳이 털리면 다른 곳도 위험해집니다.
- **MOVEit(2023)** - 최근 사례. 한 파일 전송 프로그램의 허점으로 **2,700개 이상 기관, 9천만 명 이상** 의 정보가 한꺼번에 노출됐습니다(집계가 계속 늘어난 사례입니다).

공통점이 보이나요? **입력을 함부로 믿지 않기, 비밀번호는 해시로, 알려진 취약점은 방치하지 않기** - 오늘 배운 것들이 그대로 교훈입니다. 큰 회사도 기본을 놓쳐서 뚫립니다. (출처: Breachsense, "Data Breach Examples"; Malwarebytes, "What is SQL injection")

### ② 환경변수 - 배포 전에 비밀값부터 분리하기

API 키와 DB 접속 정보는 코드에 적지 않고 환경변수로 분리합니다. 실제 유출 사례와 설정 예시는 뒤의 '보안 선택 읽기'에서 이어서 읽습니다. 배포 실습에서도 `.env`를 저장소에 올리지 않고, 배포 플랫폼에 필요한 비밀값을 따로 설정하세요.

### ③ 인증 vs 인가 - 표 검사원과 VIP 바운서

로그인 기능에는 사실 **두 단계** 가 숨어 있습니다. 콘서트장을 떠올리면 딱 맞습니다.

- **인증(Authentication)** = 입구의 **표 검사원**. "당신, 표 주인 맞아요?" → 로그인 확인.
- **인가(Authorization)** = VIP 구역의 **바운서**. "들어는 왔는데, 이 방에 들어갈 **권한** 은 있어요?"

이 두 단계에서 실패했을 때의 응답은 S2 '네트워크 기본기'의 ②에서 본 상태 코드 **401** 과 **403** 에 각각 대응합니다.

로그인에 성공하면 서버가 **토큰** 을 줍니다. 토큰은 클럽 입구에서 받는 **손목 밴드** 같은 것입니다 - 한 번 확인받고 밴드를 차면, 이후엔 매번 신분증을 안 보여줘도 밴드만 보이면 통과되죠. 브라우저는 이 밴드(토큰)를 요청마다 함께 보냅니다.

또 하나 중요한 것 - **비밀번호는 절대 그대로 저장하면 안 됩니다.** 대신 **해시(hash)** 라는, **한쪽으로만 계산되는 방식** 으로 바꿔 저장합니다. 달걀을 프라이로 만들 수는 있어도 프라이를 다시 달걀로 되돌릴 수 없듯, 해시된 비밀번호는 원래대로 되돌릴 수 없습니다. 해시는 원문이 그대로 드러나는 위험을 크게 줄이지만, 유출된 해시에 대한 추측 공격까지 없애 주지는 않아요.

AI가 이 로직을 만들어 주지만, **"내 데이터를 남이 보면 안 되는데, 권한 확인(인가)이 되어 있나?"** 와 **"비밀번호는 해시로 저장돼 있나?"** 는 꼭 되물어야 합니다. (출처: Sashido)

### 보안 이야기: 비밀번호는 어떻게 지켜지나 - 해시와 솔트

앞에서 비밀번호는 '해시'로 저장한다고 했죠. 조금만 더 들어가 봅시다. 야후(2012)처럼 **평문으로 저장했다 통째로 털린** 사고가 실제로 있었으니까요.

- **해시** 는 비밀번호를 '되돌릴 수 없는 지문'으로 바꿉니다. 같은 비밀번호는 같은 지문이 나오죠. 서버는 원문 대신 지문만 저장하고, 로그인할 때 지문끼리 비교합니다.
- 그런데 사람들이 흔히 쓰는 비밀번호(`123456`)는 지문도 뻔합니다. 그래서 **솔트(salt)** - 사람마다 다른 무작위 값 - 를 섞어 지문을 냅니다. ('소금 한 꼬집'처럼 뿌린다고 해서 붙은 이름인데, 우리말로 '소금'이라 부르진 않고 보통 '솔트'라고 합니다.) 같은 `123456`이라도 사람마다 지문이 달라지죠.
- 좋은 서비스는 **bcrypt** 같은, 일부러 느리게 계산되는 방식을 씁니다. 느린 게 장점입니다 - 도둑이 수억 개를 대입해 보기 어려워지니까요.

처음부터 해시 알고리즘을 외울 필요는 없어요. AI에게 **"비밀번호는 솔트를 섞어 bcrypt로 해시해서 저장해줘"** 라고 요구할 줄만 알면 됩니다. 실제로 어떤 서비스는 bcrypt를 써 둔 덕에, 데이터가 유출됐어도 비밀번호는 안전했습니다. (출처: Breachsense, "Data Breach Examples")

### 작동 원리: 쿠키와 세션 - 웹은 원래 건망증이 심하다

S2 '네트워크 기본기'의 ②에서 본 요청·응답 흐름 위에서, 백엔드는 여러 요청이 같은 사용자의 것인지도 확인해야 합니다. 로그인 한 번으로 다음 페이지에서도 나를 알아보게 하려면 어떻게 해야 할까요? 여기서 **쿠키** 와 **세션** 이 등장합니다.

- **쿠키 = 내 지갑 속 작은 카드.** 사이트가 번호를 하나 적어 주면, 브라우저가 보관했다가 **방문할 때마다 그 카드를 보여** 줍니다.
- **세션 = 사이트 금고 속 내 서류철.** 사이트는 카드의 번호를 보고 "아, 이 사람!" 하며 금고에서 내 정보(장바구니·권한 등)를 꺼냅니다.

로그인 흐름은 이렇습니다 - ① 로그인하면 서버가 **세션을 만들고 고유 번호(세션 ID)** 를 발급 → ② 그 번호를 **쿠키로** 브라우저에 줌 → ③ 이후 요청마다 브라우저가 그 쿠키를 함께 보냄 → ④ 서버가 번호로 나를 알아봄. 앞서 본 '손목 밴드' 토큰과 똑같은 원리죠.

> 그래서 **진짜 정보는 브라우저(쿠키)에 두면 안 됩니다.** 쿠키는 사용자가 조작할 수 있어서, `이름=일반사용자`를 `이름=관리자`로 바꾸는 장난이 가능하거든요. 그래서 브라우저엔 **의미 없는 번호만**, 진짜 정보는 **서버 금고** 에 둡니다. ('로그인 유지' 체크박스는 이 카드를 더 오래 보관하는 것입니다.) (출처: DesignGurus, "Cookies vs Sessions 101")

### 잠깐 판단 - 이 습관, 안전할까?

아래 습관이 안전하면 O, 위험하면 X를 마음속으로 매겨 보세요. 이유도 생각한 뒤 '답 확인'에서 비교해 보세요.

- API 키를 코드에 적어 GitHub에 올린다
- 사용자가 입력한 값을 그대로 믿고 처리한다
- 비밀번호를 솔트 섞어 해시로 저장한다
- 모든 사이트에 같은 비밀번호를 쓴다
- 로그인 뒤 '권한(인가)'까지 확인한다

<details class="lms-extra">

<summary>답 확인</summary>

| 습관 | 답과 이유 |
|---|---|
| API 키를 코드에 적어 GitHub에 올린다 | X(위험): 키가 유출될 수 있으니 **환경변수** 로 빼야 해요. |
| 사용자가 입력한 값을 그대로 믿고 처리한다 | X(위험): 주입 공격이 들어올 수 있어요. |
| 비밀번호를 솔트 섞어 해시로 저장한다 | O(안전): 비밀번호를 원문 그대로 저장하지 않아요. |
| 모든 사이트에 같은 비밀번호를 쓴다 | X(위험): 마이스페이스 사례처럼 한 곳에서 유출되면 다른 곳도 위험해요. |
| 로그인 뒤 '권한(인가)'까지 확인한다 | O(안전): 인증뿐 아니라 해당 작업을 할 권한도 확인해요. |

오늘 배운 것들이 그대로 정답입니다. 보안은 어려운 게 아니라 **이 습관들을 AI에게 요구하는 것** 입니다.

</details>

> **정리 - AI는 아주 빠른 신입 개발자입니다.** 속도는 놀랍지만, 보안·비밀 관리·권한 확인은 감독이 필요 합니다. 신입에게 "결제 로직을 알아서 안전하게 짜 둬"라고 방치하지 않듯이요. 위 세 가지 - 입력 검증 · 비밀은 환경변수 · 권한 확인(+비밀번호 해시) - 을 프롬프트로 요구하고, 결과를 한 번 더 되묻는 것. 그것이 처음 배우는 여러분도 할 수 있는 가장 강력한 품질 관리입니다.

## 배포 실습 1·2 - 배포하고 무료 서버 고르기

## 내 손을 떠나 세상으로 - 배포 한 번 해보기

---

**배포(deploy)** 는 내 컴퓨터에만 있던 서비스를 늘 켜진 서버에 올려, 누구나 주소로 들어올 수 있게 하는 것입니다. 처음의 그 장면 - 친구가 못 열던 화면 - 을 세상 누구나 열 수 있게 만드는 단계죠.

예전엔 서버 사고, 설정하고, 며칠씩 걸리던 전문가의 일이었습니다. 지금은 처음 배우는 사람도 **클릭 몇 번 + AI의 도움** 으로 몇 분이면 됩니다. 그 흐름을 따라가 봅시다.

배포할 곳을 고르기 전에, 서비스를 맡길 서버와 호스팅부터 살펴보겠습니다.

### 그 주소엔 무엇이 있나 - 서버와 호스팅

**서버** 는 내 서비스 파일과 데이터를 올려 두고 **24시간 켜져 있는 컴퓨터** 입니다. 내 컴퓨터에서 돌던 앱이 노트북을 닫으면 꺼지는 이유가 여기 있습니다 - 내 노트북은 24시간 켜 두는 물건이 아니니까요.

그래서 **호스팅** - 늘 켜져 있는 컴퓨터를 빌려 쓰는 것 - 을 씁니다. "그럼 나도 서버 컴퓨터를 사야 하나?" 아닙니다. **요즘은 서버를 직접 사거나 관리하지 않습니다.**

- 예전: 회사가 **서버 컴퓨터를 직접 구입** 해서, 전원·냉방·보안·고장까지 다 챙겼습니다.
- 지금: **클라우드** - 아마존·구글 같은 회사의 거대한 서버 창고에서 필요한 만큼만 빌려 씁니다. 고장·보안·전원은 그쪽이 알아서 합니다.

여러분이 할 일은 "어디에 올릴지"만 고르는 것입니다. 클릭 몇 번이면 24시간 켜진 서버가 생깁니다. (출처: Sashido, "Vibe Coding to Production"; Nucamp, "Deploying Full-Stack Apps in 2026")

![서버 랙이 줄지어 선 데이터센터 내부](https://commons.wikimedia.org/wiki/Special:FilePath/Datacenter-telecom.jpg?width=640)
*'남의 컴퓨터를 빌린다'는 건 이런 곳의 한 조각을 빌리는 것입니다 - 24시간 켜진 서버들의 방. (이미지: Wikimedia Commons)*

이제 이 호스팅 환경에서 화면·서버·데이터 중 어떤 역할을 맡길지 정해 보겠습니다.

### 화면·서버·데이터, 어디에 맡기면 될까요?

화면·서버·데이터 저장소는 서로 다른 **역할** 을 맡아요. 배포할 때 꼭 세 군데에 나눠 올려야 하는 것은 아니에요. 화면과 서버를 한 플랫폼에 함께 올릴 수도 있고, 필요한 기능만 다른 서비스에 맡길 수도 있어요.

| 필요한 역할 | 이번 수업의 선택 | 무엇을 맡기나요? |
|---|---|---|
| **화면**(프론트) | GitHub Pages · Vercel · Netlify | 브라우저에 보여 줄 화면을 공개해요. 다음 표에서 내 프로젝트에 맞게 골라요. |
| **데이터 저장·로그인** | **Supabase**(Todo 앱의 데이터 저장 서비스) | Todo 앱의 데이터 저장·재조회 연결을 그대로 유지해요. 이미 준비된 백엔드 기능을 연결해서 쓰는 방식을 BaaS라고 해요. |
| **직접 만든 서버**(대안) | Railway · Render | 내가 작성한 API와 처리 로직을 실행해요. 직접 서버 코드가 필요할 때 골라요. |

오늘 기본 경로는 **Todo 앱에 연결된 Supabase를 그대로 유지하고 화면을 배포하는 것** 이에요. Firebase는 다른 프로젝트에서 로그인·저장을 맡길 수 있는 대안 서비스예요. 직접 API 서버를 만들기로 했다면 **Railway나 Render** 에 배포하고, 데이터 저장은 별도로 연결할 수 있어요. 둘을 모두 써야 하는 것은 아니에요.

선택 사항인 Firebase의 **Spark 요금제** 는 결제 정보 없이 시작할 수 있고, 인증의 여러 방식과 Firestore 데이터 저장을 무료 범위 안에서 쓸 수 있어요. 무제한은 아니며 기능마다 한도가 달라요. 직접 서버 코드를 실행하는 Cloud Functions는 Blaze 요금제 연결이 필요하니, Spark에서 모든 백엔드 기능이 무료라고 생각하면 안 돼요. ([Firebase 공식 요금제 안내](https://firebase.google.com/docs/projects/billing/firebase-pricing-plans))

> Railway는 시작 크레딧을 다 쓰면 유료로 넘어갑니다(뒤의 '무료 서버' 섹션에서 자세히).

![로켓이 발사되는 장면](https://commons.wikimedia.org/wiki/Special:FilePath/Rocket_launch.jpg?width=640)

### 내 코드에 맞는 곳을 고르고, GitHub와 연결해요

먼저 **정적 화면인지, 서버 코드도 실행해야 하는지** 를 확인해요. 정적 화면은 미리 만들어 둔 HTML·CSS·JavaScript를 보내면 되지만, 풀스택 앱은 화면과 함께 서버 쪽 처리도 필요해요.

| 플랫폼 | 무엇이 다른가요? | 이럴 때 골라요 |
|---|---|---|
| **GitHub Pages** | 정적 파일을 공개하는 곳이에요. 직접 만든 API 서버를 실행하지는 못해요. GitHub Free에서는 공개 저장소로 이용해요. | 소개 페이지·포트폴리오처럼 화면만 있고, 소스 공개가 괜찮을 때 |
| **Vercel** | 정적 화면뿐 아니라 Next.js의 서버 기능도 배포할 수 있어요. | Next.js로 화면과 서버 기능을 함께 만든 경우의 기본 후보 |
| **Netlify** | 정적 사이트와 지원하는 프레임워크의 서버 기능을 배포할 수 있어요. Next.js도 지원하지만 지원 범위·설정은 Vercel과 같지 않아요. | 정적 화면을 GitHub에 연결해 배포하거나, 사용하는 프레임워크의 Netlify 안내가 잘 맞을 때 |
| **Railway / Render** | 직접 만든 백엔드 앱의 실행 환경을 맡기는 곳이에요. | Express·FastAPI 등으로 별도 API 서버를 만들고, 시작 명령으로 실행해야 할 때 |

**화면만 있다면 GitHub Pages나 Netlify, Next.js의 서버 기능까지 쓴다면 Vercel, 별도 API 서버라면 Railway/Render** 를 먼저 살펴봐요. 오늘은 Todo 앱에 연결된 **Supabase** 를 그대로 사용해요. 다른 프로젝트에서 Firebase를 연결하는 것은 선택 사항이에요. 정적 화면에서도 외부 백엔드를 호출할 수 있으므로, 로그인 화면이 있다고 무조건 풀스택 배포를 골라야 하는 건 아니에요.

가입 전에는 무료 한도·상업적 이용 조건과 내 프레임워크 지원 범위를 확인해요. GitHub Pages의 비공개 저장소 지원은 유료 플랜 등에 따라 달라요. **저장소가 비공개라고 배포한 사이트까지 비공개인 것은 아니에요.**

공식 안내: [GitHub Pages](https://docs.github.com/en/pages/getting-started-with-github-pages/about-github-pages) · [Vercel의 Next.js 지원](https://vercel.com/docs/frameworks/full-stack/nextjs) · [Netlify의 Next.js 지원](https://docs.netlify.com/build/frameworks/framework-setup-guides/nextjs/overview/)

#### 처음 연결할 때

1. 비밀키·`.env`가 포함되지 않았는지 확인하고, 프로젝트를 **커밋** 해요. 커밋은 변경 내용을 하나의 기록으로 남기는 일이에요.
2. **push** 해서 그 기록을 GitHub 저장소로 보내요.
3. 플랫폼에서 저장소를 연결하고 **배포할 브랜치** 와 빌드·실행 설정을 확인해요. GitHub Pages는 저장소의 Pages 설정에서 게시 방식을 정해요.
4. 첫 배포가 성공하면 제공된 **공개 주소** 로 접속해 확인해요. Vercel이라면 `내프로젝트.vercel.app` 같은 주소예요.

#### 코드를 고친 다음부터

1. 코드를 **수정하고 저장** 해요. 이 단계에서는 내 컴퓨터만 바뀌어요.
2. 바뀐 파일을 확인하고 **커밋** 해요.
3. 연결한 배포 브랜치로 **push** 해요.
4. 자동 배포를 켜 둔 플랫폼이 새 커밋을 받아 **빌드·배포** 해요. 실패했다면 로그를 확인해요.
5. 배포 성공 표시를 확인한 뒤 **공개 URL에서 변경 내용을 확인** 해요.

**저장만으로는 공개 서비스가 바뀌지 않아요.** GitHub에 연결된 배포 브랜치에 변경 기록이 도착하고 배포까지 성공해야 반영돼요.

> 그리고 이 순간 - 여는 이야기에서 친구에게 보냈지만 **안 열리던 그 링크** 대신, 이제 `내프로젝트.vercel.app` 같은 공개 주소를 보낼 수 있어요. 화면만 있던 가게에 마침내 문이 열린 거예요.

### 배포에서 초보가 가장 많이 막히는 세 곳

이 세 가지만 미리 알면, "왜 안 되지?"의 대부분을 피할 수 있습니다.

- **① 환경변수 빼먹기** - 내 컴퓨터의 `.env`(비밀값)는 GitHub에 **일부러 안 올립니다**(올리면 유출이니까). 그래서 배포 플랫폼의 **설정 화면에 비밀값을 따로 넣어** 줘야 합니다. 이걸 빼먹어서 "로컬에선 됐는데 배포하니 안 돼"가 가장 많습니다. (출처: Nucamp 2026)
- **② CORS 막힘** - 화면(Vercel)과 서버(Railway)를 연결할 때 CORS 오류가 나면, S6 '프론트-백엔드 연결'의 CORS 설명과 허용 설정을 확인하세요.
- **③ 무료의 함정** - 대부분 무료 티어가 있지만(예: Vercel 무료, Railway 시작 크레딧, Firebase 무료 한도 - 2026년 기준, 정책은 바뀔 수 있음), **트래픽이 늘면 유료로 넘어갑니다.**

### 요금 폭탄 피하기

무료로 시작해도, 방심하면 요금이 훅 올라갈 수 있습니다. 특히 앞에서 본 **키 유출** 이나, 실수로 만든 **무한 반복** 때문에요. 초보가 지킬 세 가지.

- **요금 알림을 켜 둔다** - "한 달에 X원 넘으면 알려줘"를 설정해 두면 급증을 알아차리는 데 도움이 되지만 자동으로 차단해 주는 상한은 아니에요.
- **한도(예: 월 무료 사용량)를 안다** - 내가 쓰는 플랫폼의 무료 한도가 얼마인지 확인해 둡니다.
- **비밀 키를 지킨다** - 유출된 키로 남이 내 계정에서 비싼 작업을 돌리면 그 요금은 내 몫입니다. (그래서 환경변수가 중요했죠.)

> 이건 겁주려는 게 아닙니다. 알고 시작하면 **거의 다 예방 가능한** 것들입니다.

### AI에게 배포를 부탁하는 프롬프트

문법을 몰라도, **조건 + 순서 + 함정 체크** 를 요구하면 AI가 처음부터 끝까지 안내자 역할을 합니다.

```text
이 프로젝트를 배포하고 싶어. 비용은 최소로 하고 싶어.
1. 프론트엔드와 백엔드를 각각 어디에 올리면 좋을지 무료 옵션으로 추천해줘.
2. GitHub에 올리고 자동 배포되게 하는 순서를 하나씩 알려줘.
3. 비밀 키(환경변수)는 어디에 어떻게 넣어야 하는지 짚어줘.
4. 배포 후 화면이 서버와 통신이 안 될 때 확인할 것(CORS·환경변수)을 체크리스트로 줘.
5. 요금이 갑자기 오르지 않게 설정할 것이 있으면 알려줘.
```

## 공짜로 세상에 올리기 - 무료 서버 고르기

앞에서 "배포"를 배웠죠. 그럼 자연스러운 질문 - **돈 안 들이고 올릴 수 있나요?** 네, 됩니다. 요즘은 처음 만든 웹앱도 무료로 인터넷에 띄울 수 있는 곳이 여럿입니다.

다만 한 가지 재미있는 흐름을 먼저 알아 두면 좋습니다. **공짜의 세계는 계속 줄어드는 중** 입니다.

- **Heroku** - 한때 초보들의 성지였던 무료 서버가 **2022년 11월 완전히 사라졌습니다.**
- **Railway** - 2023년, **Fly.io** - 2024년에 상시 무료를 없앴습니다.
- **Glitch** - 2025년 7월 앱 호스팅을 종료했습니다.

왜일까요? 무료 계정을 **암호화폐 채굴** 같은 곳에 악용하는 사람이 늘면서, 회사들이 공짜를 거둬들인 겁니다. (출처: Heroku 공식 FAQ; Railway Blog) 그러니 "지금 무료인 것도 언젠가 바뀔 수 있다" 는 전제로, 아래 목록도 가입 직전에 공식 요금 페이지를 한 번 확인 하는 습관이 안전합니다.

### 무료 티어, 시작 전에 알아 둘 세 가지

무료 서비스도 조건을 알고 쓰면 실습과 시연에 도움이 돼요. 특히 **서버가 깨어 있는지** 와 데이터가 어디에 저장되는지 는 따로 확인해요.

1. **잠듦(spin-down)** - Render 무료 웹 서비스는 들어오는 요청·메시지가 **15분 동안 없으면 잠들어요.** 다음 요청이 오면 다시 시작하며 약 1분 걸릴 수 있어요. 첫 접속이 느리다고 바로 고장은 아니에요.
2. **데이터 보관** - Render 무료 웹 서비스의 로컬 파일·SQLite 변경 내용은 재시작·재배포·절전 때 사라져요. 오래 보관할 데이터는 Firebase 같은 외부 데이터 저장소에 연결해요. Render 무료 Postgres는 별도 서비스이며 생성 후 30일에 만료 돼요. 중요한 데이터는 백업하고, Firebase도 사용하는 제품의 보관·한도 정책을 공식 문서에서 확인하세요.
3. **사용량과 요금** - Firebase Spark는 무료 한도를 넘기면 해당 기능 사용이 제한될 수 있어요. 유료 사용량에 따라 청구하는 요금제로 바꾼다면 사용량을 확인하고 예산 알림을 설정해요. **예산 알림은 요금을 자동으로 막아 주는 상한이 아니에요.** 자동 제한 기능이 있는지는 서비스별로 확인하세요.

공식 안내: [Render 무료 서비스의 절전·파일 보관·만료 조건](https://render.com/docs/free) · [Firebase 요금제와 사용량 제한](https://firebase.google.com/docs/projects/billing/firebase-pricing-plans)

> 정리하면 - **무료는 "배우고 시연하기"엔 충분** 합니다. "많은 사람에게 안정적으로 24시간 제공"하는 단계가 오면, 그때 유료를 검토하면 됩니다.

## 배포 실습 3 - 내 컴퓨터를 서버로

## 내 컴퓨터를 서버로 - 집에서 돌려보기 (선택 실습)

"굳이 남의 서버를 빌려? 내 컴퓨터를 켜 두면 되잖아?" - 맞는 생각이고, 실제로 됩니다. 다만 왜 **기본적으로는 안 되는지** 부터 알면 훨씬 수월합니다.

S2 '네트워크 기본기'의 ①에서 본 사설 IP·NAT를 떠올리면, 집 안에서 열리는 서버를 외부에도 공개하려면 별도의 연결 설정이 필요하다는 점을 이해할 수 있습니다.

잠깐 시연하려면 아래의 터널 도구를 살펴보세요. 직접 집 서버를 운영하고 싶다면 가정 회선의 조건부터 확인해야 합니다.

<details class="lms-extra">

<summary>선택 읽기: 가정 회선에서 먼저 확인할 것</summary>

집 안 전용 주소(사설 IP, 192.168.…)와 별개로, 집 회선의 바깥 주소는 대개 고정이 아니라 통신사가 수시로 바꾸는 **유동 IP** 입니다.

전통적인 연결 설정은 두 가지입니다.

- ① **포트 포워딩**: 공유기에 외부 요청을 내 서버 컴퓨터로 전달하도록 설정합니다.
- ② **동적 DNS**: 계속 바뀌는 집 주소를 같은 이름으로 찾아갈 수 있게 합니다.

요즘은 이 설정 부담을 줄여 주는 터널 도구도 있습니다.

</details>

### 전통 방법: 포트 포워딩 + DDNS

S2 '네트워크 기본기'의 ①에서 본 포트를 기준으로, 직접 집 서버를 공개할 때의 설정과 주의점을 살펴봅니다.

<details class="lms-extra">

<summary>선택 읽기: 포트 포워딩의 노출 위험과 가정 회선 제약</summary>

공유기의 포트 포워딩과 DDNS 설정을 직접 해 보면 연결 원리를 익힐 수 있습니다.

다만 다음 위험을 주의해야 해요.

- **집 네트워크가 인터넷에 직접 노출** 됩니다. 열어 둔 문으로 전 세계 자동 공격 프로그램이 문을 두드립니다.
- 요즘 많은 가정 회선은 통신사가 주소 하나를 여러 집이 나눠 쓰는 방식(CGNAT) 이라, 포트 포워딩 자체가 아예 안 되는 경우가 많습니다.

</details>

### 요즘 편한 무료 방법 - '터널'로 문 없이 연결하기

집 서버를 잠깐 공개할 때는 터널 도구를 이용할 수 있습니다. 아래에서 공개 범위와 무료 한도를 비교해 보세요.

<details class="lms-extra">

<summary>선택 읽기: 터널은 어떻게 집 서버에 연결하나요?</summary>

터널(tunnel) 은 밖에서 우리 집으로 들어오게 여는 대신, 내 컴퓨터가 바깥 서비스로 먼저 연결을 맺어 그 통로로 손님을 받는 방식입니다. 그래서 포트를 열 필요가 없고, CGNAT도 통과 합니다.

</details>

|---|---|---|---|
| **Cloudflare Tunnel** | 포트 개방 불필요, 정식 주소·인증서까지 | 무료로 시작할 수 있지만, 용도에 맞는 이용 조건과 초기 설정 확인 필요 | 내 컴퓨터의 서버를 외부와 연결할 때 |
| ngrok | 한 줄 명령으로 공개 주소 연결 | 무료는 계정에 할당된 개발용 주소 1개·동시에 온라인인 엔드포인트 최대 3개. HTTP/S 접속에 안내 페이지가 표시되며 사용량 한도도 있음 | 잠깐 시연·테스트 |
| **Tailscale** | 팀 내부 공유와 외부 공개 모두 가능 | 내부 공유는 Serve, 공개 인터넷에 열 때는 Funnel. 무료 플랜 조건과 공개 범위 확인 필요 | 내부 공유 또는 잠깐 공개 시연 |

내 컴퓨터에서 도는 화면을 잠깐 친구에게 보여줄 때는 **ngrok** 같은 터널이 편해요. 다만 무료가 무제한이라는 뜻은 아니에요. ngrok의 현재 요금 페이지에는 일회성 사용 크레딧과 사용량 한도가 안내되어 있으니, 시작 전에 남은 한도를 확인하세요. Tailscale도 **Funnel로 열면 팀 밖의 인터넷 이용자가 접속할 수 있다** 는 점을 기억하세요. (출처: ngrok 요금 페이지; Tailscale Funnel·Serve 공식 문서)

> Todo 앱의 데이터는 Supabase에 저장돼요. 오늘은 그 화면을 배포하는 게 기본 경로예요. 직접 만든 서버 코드를 배포하고 싶다면 Railway/Render 가 대안이고, 터널은 내 컴퓨터에서 도는 것을 그대로 잠깐 보여주고 싶을 때 고르면 돼요.

## 배포 실습 4 - 프론트와 백엔드 연결하기

## Todo 앱을 배포 환경에서 확인하기

이 실습에서는 Supabase에 연결해 할 일을 저장하고 다시 불러올 수 있는 Todo 앱을 사용해요. 오늘은 그 연결을 유지한 채, 친구도 접속할 수 있는 공개 화면에서 같은 기능이 동작하는지 확인해요.

오늘 실습의 기본 순서는 이래요.

1. **기존 Supabase 연결 유지** - Supabase로 저장·재조회가 되는 Todo 앱을 그대로 사용해요.
2. **배포 환경에서 환경변수 설정** - 배포 서비스에 필요한 값을 설정해요. 브라우저에 공개할 수 있는 값과 서버에서만 쓸 비밀값을 구분하고, 비밀값은 화면 코드에 넣지 않아요.
3. **공개된 화면에서 요청·저장·재조회 확인** - 공개 주소에 접속해 할 일을 저장하고, 새로고침한 뒤에도 다시 조회되는지 확인해요. 네트워크 탭에서 요청과 응답도 살펴봐요.

<details class="lms-extra">

<summary>선택 읽기: 아직 백엔드를 연결하지 않은 다른 프로젝트라면</summary>

Todo 앱에 Supabase 연결이 되어 있다면 이 보충 경로는 필요하지 않아요.

- **목업 화면 먼저** - 가짜 데이터로 화면 동작을 확인해요.
- **백엔드 API 붙이기** - 화면이 호출할 창구(엔드포인트)를 만들어요.
- **가짜 데이터를 실제 호출로 교체** - 목업 자리를 서버 응답으로 바꿔요. 연결을 마치면 위의 배포 환경 점검으로 돌아와요.

</details>

> 한 가지 결정 - **"앱 전체를 한 번에 시킬까, 나눠 시킬까?"** 통째로 한 프롬프트에 다 시키는 건 초보에게 거의 항상 실패합니다(AI의 집중력·기억 한계). **"데이터 모양 → API → 화면"** 처럼 작게 쪼개 시키면 실수를 일찍 잡습니다. (출처: Red Hat Developer, "Big vs. small prompts"; Addy Osmani)

배포 환경에서 동작을 확인한 뒤에는 오늘의 실습 결과물을 확인해요.

- **백엔드가 실제로 돌고 공개 URL로 접속되는 서비스**: 내 컴퓨터뿐 아니라 친구의 브라우저에서도 열고, 백엔드 기능이 동작하는지 확인해요.

### 실전 팁: 연결에서 초보가 실제로 겪는 문제들

연결 단계는 실무자들이 "바이브가 죽는 순간"이라 부를 만큼 문제가 몰립니다. 대표적인 것만 - **증상 → 처방** 으로.

- **계약 불일치** (가장 큰 원흉) - 프론트가 기대하는 데이터 모양 ≠ 백엔드가 주는 모양. AI를 쓰면 오히려 심해집니다(양쪽에 '얼추 맞는' 걸 따로 만들어 서서히 어긋남). *증상:* 화면에 `undefined`가 뜨거나, 기대와 달리 한 겹 더 감싸인 응답이 옴. *처방:* '나중에 맞추기'가 아니라 **주고받을 데이터를 처음부터 문서로 적어 두고**(API 명세) 그걸 기준으로 양쪽을 짭니다 - 아래 전략에서 자세히.
- **로딩·에러 상태 누락** - 데이터가 오기 전 화면이 깜빡이거나, 서버가 오류를 줘도 앱이 멈춘 듯 보임. *처방:* 모든 호출을 **로딩 / 성공 / 에러** 세 상태로 나눠 각각 화면을 준비.
- **CORS·주소·환경변수** - 연결이 막히면 바로 아래 '실전 팁: 연결을 매끄럽게 하는 프롬프트 전략'의 CORS 설명과 점검 요청을 확인하세요.

(출처: Darren Coxon, "Golden Rules of Full-Stack Vibe"; c-sharpcorner)

### 실전 팁: 연결을 매끄럽게 하는 프롬프트 전략

화면과 API를 서로 다른 주소에 배포했다면 **CORS** 설정도 확인해야 합니다. 브라우저는 출처(프로토콜·호스트·포트의 조합)가 다른 서버의 응답을 화면 코드가 읽을 때, 서버가 그 출처를 허용했는지 확인합니다. 이 허용 정보를 주고받는 방식이 CORS입니다. 화면은 열리는데 API 응답을 읽지 못하고 개발자도구에 CORS 오류가 나오면, AI에게 *"백엔드가 내 프론트 주소를 CORS에서 허용하는지 확인하고, API 주소는 환경변수로 관리해 줘."* 라고 요청하세요.

가장 강력한 전략은 주고받을 데이터를 '문서'로 남기고 시작하는 것 입니다. '먼저 정한다'에서 끝내지 말고, 아예 문서로 적어 두면 계약 불일치가 근본적으로 사라집니다.

- API 명세를 문서로 (근본 해결) - 어떤 창구(엔드포인트)가 있고, 각 요청·응답이 어떤 JSON 모양 인지 적은 문서를 만들어 두세요. AI에게 이렇게: *"엔드포인트 목록과 각 요청/응답 예시(JSON)를 명세 문서로 만들어 줘. 앞으로 이 문서를 기준으로 프론트와 백엔드를 짜자."* 기능이 바뀌면 코드보다 이 문서를 먼저 고칩니다. 프론트·백엔드가 늘 같은 문서를 보면 서로 어긋날 일이 없습니다. (형식을 갖추려면 OpenAPI 같은 표준도 있지만, '이름표-값' 예시를 적은 간단한 메모라도 충분합니다.) (출처: Evil Martians, "API contracts survival guide")
- **한 기능씩 끝까지(수직 슬라이스)** - "화면 전부 → 서버 전부"로 층을 나누면 AI가 다른 층을 몰라 추측하다 틀립니다. 대신 **"회원가입 한 기능"을 화면-API-저장까지 하나로** 완성하고 바로 테스트하세요.
- 양쪽을 함께 보게 - 프론트와 백엔드 코드를 같은 대화·같은 저장소 에 두면 AI가 양쪽 모양을 맞추기 수월합니다.
- **목업→실제 전환** - *"지금 화면이 쓰는 가짜 데이터를 방금 만든 실제 API 호출로 바꿔 줘. 문서(명세)의 요청/응답 모양은 그대로 유지하고, 로딩과 에러 처리도 함께 넣어 줘."*
- **매 단계 커밋** - 되돌아갈 체크포인트를 만들어 두세요. AI는 확신에 차서 멀쩡한 걸 망가뜨리기도 합니다.

### 실전 팁: AI 풀스택의 단골 실수 + 부담 줄이는 스타터

AI는 보안이 나빠서가 아니라 **프롬프트에 보안이 없었기 때문에** "일단 돌아가는" 코드를 만듭니다(목표가 '작동'이지 '안전'이 아니라서). 내보내기 전 이것만은 점검하세요.

- **프론트에만 검증, 백엔드 검증 누락** - 공격자는 화면을 안 거치고 API를 직접 부를 수 있습니다. 서버에서 입력을 다시 검증해야 합니다(앞서 본 "프론트는 못 믿는다").
- **하드코딩된 주소 / 키 노출** - S2 '네트워크 기본기'의 ①에서 본 `localhost`를 떠올리며, 배포용 API 주소가 환경변수에 들어 있는지 확인하세요. API 키를 프론트 코드에 넣으면 개발자도구로 다 보입니다. (특히 `NEXT_PUBLIC_`·`VITE_` 접두사가 붙은 값은 "브라우저에 공개"라는 뜻이니 비밀값엔 절대 쓰지 마세요.)
- **검증 팁** - 배포 전 AI에게 한 번 시키세요: *"이 코드 전체를 하드코딩된 비밀값·서버 입력 검증 누락·CORS 설정·인증 취약점 관점으로 점검해 줘."* (출처: Zuplo; Wiz)

> 오늘은 Todo 앱에 연결된 Supabase를 유지해요. 아직 백엔드를 연결하지 않은 다른 프로젝트에서 연결 부담을 줄이고 싶다면, 선택 사항으로 **Firebase로 데이터 저장·인증을 연결하는 예제** 를 참고해도 좋아요. 다만 예제가 돌아간다고 보안까지 확인된 것은 아니에요. 누가 어떤 데이터를 읽고 바꿀 수 있는지 접근 규칙을 정하고, 다른 사용자로도 시험해 보세요. 편리한 도구를 써도 점검은 여전히 사람의 몫입니다.

흐름을 정리해 볼게요. Todo 앱은 **Supabase에 연결해 데이터를 저장하고 다시 불러와요**. 오늘은 그 연결을 유지하고, 배포 환경에서 필요한 환경변수를 설정했어요.

오늘 배운 네트워크(연결) · 보안(안전) · 배포(공개)는 Todo 앱을 **안전하게 세상에 공개하기** 위한 준비였어요. 공개된 화면에서도 요청과 응답을 살펴보고, 저장한 할 일이 새로고침 뒤에도 다시 조회되는지 확인해요.

이제 맨 처음의 그 장면처럼 **친구도 링크를 열 수 있고, 새로고침해도 데이터가 남는 것** 을 확인할 수 있어요. 내 컴퓨터에서만 열리던 Todo 앱에 친구도 접속할 수 있게 됐어요.

## 보안 선택 읽기 - 환경변수와 설치·패치 습관

### 환경변수 - 비밀은 코드에 적지 않는다

가장 흔한 초보 사고를 이야기로 봅시다.

> 한 개발자가 급한 마음에 API 키를 **코드에 그대로 적어** GitHub에 올렸습니다. "잠깐인데 뭐 어때." 그런데 **몇 분 만에** 자동 프로그램(봇)이 그 키를 발견해 훔쳐 갔고, 그 키로 클라우드에서 **비싼 작업을 잔뜩 돌려** 며칠 뒤 수백만 원짜리 요금 청구서가 날아왔습니다.

과장이 아닙니다. 보안 회사 GitGuardian 집계로, **2025년 한 해 공개 GitHub에 유출된 비밀키·비밀번호가 약 2,865만 개** 로, 전년 대비 34% 늘어 역대 최대 규모였습니다. 특히 **AI 서비스 키 유출이 1년 새 81% 급증** 했고, 과거 유출된 것의 **상당수가 아직도 살아 있어** 악용됩니다. (출처: GitGuardian, "State of Secrets Sprawl 2026")

해결책이 **환경변수** 입니다. 비밀값을 **코드 바깥에 따로** 두고, 실행할 때만 꺼내 쓰는 방식이죠.

```text
나쁜 예 (코드에 그대로 박아 넣음):
   apiKey = "sk-a1b2c3d4e5f6..."      ← 이대로 GitHub에 올라가면 순식간에 유출

좋은 예 (코드는 '불러오기'만):
   apiKey = 환경변수에서 불러오기       ← 실제 값은 코드 밖 .env 파일에 보관
   (그리고 .env 파일은 깃에 안 올라가게 .gitignore로 막아 둠)
```

AI에게 이렇게 말하면 됩니다 - "API 키와 DB 접속 정보는 코드에 넣지 말고 **환경변수로 빼줘.** `.env` 파일은 깃에 안 올라가게 해줘." 한 줄 요약: **코드 = 공개해도 되는 것 / 환경변수 = 절대 공개하면 안 되는 것(비밀번호·키).**

2026년 9월 Next.js에서는 동적 이미지 생성(`next/og`)에 사용자 입력을 직접 넣을 때 원격 코드 실행이 가능한 문제가 발견됐어요. Node.js용 구현의 `next` 16.2.0～16.3.5가 대상이에요.
제공된 예제에서 해당 기능을 사용하지 않는 것이 확인된 경우에만 이 취약점의 적용 대상이 아니라고 판단할 수 있어요. 각자 수정한 프로젝트는 사용하는 기능·버전·노출 조건을 따로 확인해야 해요. 만약 내 프로젝트가 이 기능을 쓰고 있다면 미루지 않고 그날 바로 16.3.6 이상으로 올려요. ([Next.js 공식 공지](https://github.com/vercel/next.js/security/advisories/GHSA-vcvr-r3jv-pc5j))
2026년 10월 Node.js도 보안 업데이트를 냈어요. 대표적으로 공격자가 보낸 Brotli 압축 응답을 풀어주는 공개 서버에서 원격 코드 실행이 가능한 문제를 고쳤어요.
오늘 배포한 프로젝트가 사용하는 기능·버전·노출 조건을 확인하세요. 이미 공개한 서비스이니 확인을 미루면 안 돼요. 공지가 나온 날 바로 내 프로젝트가 조건에 들어가는지 확인하고, 들어간다면 공지의 수정 버전으로 올리는 게 평소 습관이에요. ([Node.js 공식 공지](https://nodejs.org/en/blog/vulnerability/october-2026-security-releases))
2025년 9월 Shai-Hulud는 악성 npm 패키지의 설치 스크립트가 환경변수·GitHub 토큰을 훔치고 스스로 퍼진 사건이에요. 같은 방식의 공격은 2026년에도 이어지고 있어요. ([Wiz 분석](https://www.wiz.io/blog/shai-hulud-npm-supply-chain-attack))
이건 버전보다 “무엇을 설치하느냐”의 문제예요. AI가 `npm install`을 제안하면 패키지 이름·공식 문서·설치 이유를 확인하고, 출처가 불분명하면 실행 전에 강사에게 물어보세요.

1. 뉴스에서 문제가 생긴 기능과 노출 조건을 확인하세요.
2. 내 프로젝트가 그 기능을 쓰고, 그 조건에 해당하는지 확인하세요.
3. 해당한다면 미루지 않고 공지의 수정 버전 이상으로 그날 올리세요.

확인은 특별한 날에 하는 게 아니라 공지가 나오는 날 바로 하는 습관이에요. 수업 중 버전은 운영진이 관리하지만 '내가 쓰는 부분인가' 확인하기는 언제나 여러분 몫이에요.
`.gitignore`는 Git 업로드만 막고 설치한 프로그램의 로컬 파일 접근은 막지 못해요. 공개 저장소에 올리기 전 커밋 목록에 `.env*`·토큰이 든 `.npmrc`가 없는지 확인하세요.
노출된 키는 파일에서 삭제하는 것으로 끝내지 말고 폐기 후 재발급하세요.
