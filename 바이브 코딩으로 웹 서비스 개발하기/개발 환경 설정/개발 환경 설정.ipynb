## AI에게 일을 맡길 작업대를 차린다는 것

![AI에게 일을 맡기는 개발 작업대](https://ebhtmardjzsiyxjncvbb.supabase.co/storage/v1/object/public/images/01-vibe-coding-web/vibe-coding-web-1-fig01.svg)

우리는 AI에게 일을 시키려면 문제를 작은 단위로 쪼개야 한다는 점을 다뤘습니다. 오늘은 그 일을 실제로 맡길 자리를 만듭니다. 머릿속 분해가 작업으로 바뀌려면, AI가 파일을 만들고 고치고 여러분이 결과를 바로 확인할 **작업대**가 필요합니다.

도구 이름이 여러 개 나오면 초반에는 산만하게 느껴질 수 있습니다. Claude Code, 터미널, Node.js, Git이 한꺼번에 등장하기 때문입니다. 이 과정에서 여러분이 실제로 일을 맡길 메인 도구는 **Claude Code**입니다.

왜 채팅만 하는 AI가 아니라 Claude Code일까요? Claude Code는 여러분이 자연어로 지시하면 **직접 파일을 만들고 고치고 명령을 실행**합니다. 웹서비스를 만들려면 답이 아니라 실제 파일이 필요하고, 그 파일이 내 컴퓨터에서 돌아가는지 확인해야 합니다. 이 '말하면 손이 움직이는' 협업이 바이브 코딩의 핵심 역량이고, 그 과정이 눈에 보여야 배움이 남습니다. 이런 도구를 보통 **에이전틱 코딩 도구**라고 부릅니다.

Claude Code는 두 모습으로 만납니다. **Claude Desktop 안의 Claude Code**(그래픽 화면)와 **터미널의 `claude` 명령**(글자 화면)입니다. 둘은 같은 엔진을 쓰는 같은 도구입니다. 왜 굳이 두 가지일까요?
![claude desktop을 활용한 claude code 화면](https://raw.githubusercontent.com/SunCreation/aiffel-contents/refs/heads/main/docs/screenshots/claude-code-in-claude-desktop.png)
- **Claude Desktop**은 마우스로 조작하는 그래픽 화면입니다. 변경내역(diff)을 시각적으로 보여주고 앱 미리보기를 띄워, 초심자가 결과를 검토하기에 부담이 적습니다. 여러 세션을 동시에 띄워 병렬로 작업할 수도 있습니다.
![claude code 를 TUI 툴로써 활용하는 모습](https://raw.githubusercontent.com/SunCreation/aiffel-contents/refs/heads/main/docs/screenshots/claude-code-cli-welcome-screen.png)
- 터미널의 `claude`는 글자로만 오가는 화면입니다. 불편해 보이지만, 개발자들이 여전히 터미널을 쓰는 이유가 있습니다. 시작이 즉각적이고, 허용된 경로의 파일을 다룰 수 있고, 스크립트로 묶어 자동화하거나 Agent SDK로 프로그램 안에 끼워 넣을 수 있습니다. Anthropic 공식 문서도 터미널을 "가장 완전한 형태의 인터페이스"라고 소개하며, 스크립팅과 Agent SDK는 터미널에서만 쓸 수 있다고 덧붙입니다.

그래서 이 과정의 예시는 두 모습 모두에서 따라갈 수 있습니다. 초심자라면 Claude Desktop으로 시작하고, 특정 자동화 같은 고급 흐름이 필요해지면 터미널로 넘어가면 됩니다. 공통으로 익힐 것은 **자연어로 지시하고 결과를 판단하는 흐름**입니다.

Cursor 같은 GUI 에디터나 opencode 같은 다른 에이전틱 도구를 쓰고 싶다면 그래도 좋습니다. 여기서 익힌 작업 방식은 도구가 바뀌어도 그대로 기준점이 됩니다. 오늘의 목표는 각 도구의 문법을 외우는 것이 아닙니다. 목표는 하나입니다. **Claude Code에 정확히 지시하고, 만들어진 결과를 내 컴퓨터에서 확인하고, 다시 수정 요청할 수 있는 흐름을 한 번 여는 것**입니다.

### 오늘의 핵심 한 문장

> **개발 환경은 AI에게 일을 시키고, 그 결과를 바로 검토·수정하기 위한 작업대입니다.**

바이브 코딩에서 중요한 일은 자연어로 만들 것을 말하고, AI가 만든 파일을 읽고, 위험하거나 어색한 부분을 찾아 다시 지시하는 과정을 반복하는 것입니다.

오늘 작업대는 다섯 가지로 이루어집니다. 각 도구가 **무엇이고, 왜 이 과정에 필요한지**를 한 줄씩만 짚고 갑니다.

| 도구 | 무엇 | 왜 이 과정에 필요한가 |
|---|---|---|
| **Claude Code** | 자연어로 지시하면 파일을 직접 만들고 고치는 에이전틱 코딩 도구 | 이 과정의 메인 도구입니다. 실제 서비스 파일을 만들고 고칩니다 |
| **Node.js와 npm** | Next.js가 돌아가는 실행 엔진(Node.js)과 그 위 패키지 매니저(npm) | 이 모듈에서 쓸 Next.js 프로젝트를 만들고 실행하려면 이 엔진이 필요합니다 |
| **uv** | 파이썬 패키지와 프로젝트를 빠르게 관리하는 도구(Astral사, Rust로 만듦) | 뒤 모듈에서 Claude가 MCP 서버나 파이썬 기반 AI 도구를 쓸 때 필요합니다. 필요할 때 설치하는 선택 도구입니다 |
| **Git** | 파일 변경 이력을 남기는 장치 | 변경을 기록해 두면 AI가 고친 내용을 비교하고 되돌릴 수 있습니다 |
| **터미널** | 글자로 명령을 내리는 창 | uv·node·git 명령과 Claude가 남기는 로그를 직접 눈으로 보는 자리입니다 |

터미널은 **명령을 입력하고 실행 결과를 읽는 창**입니다. Claude Code의 대화 입력란과 구분해서 사용합니다.

- **macOS:** `⌘ Space`를 누른 뒤 `Terminal`을 검색해 엽니다.
- **Windows:** 시작 메뉴에서 `PowerShell`을 검색해 엽니다. Windows Terminal을 쓰면 **PowerShell 프로필**을 선택합니다. 아래 Windows 명령은 PowerShell 기준이며, 명령 프롬프트(cmd)나 WSL에 섞어 입력하지 않습니다.
- `node --version` 같은 명령은 일반 터미널에, 자연어 요청은 Claude Code에 입력합니다. CLI의 Claude Code를 사용 중이라면 `/exit`로 나와 셸 프롬프트가 보일 때 명령을 입력합니다.

이해되지 않는 명령은 붙여 넣기 전에 “이 명령이 무엇을 바꾸나요?”라고 AI에게 먼저 물어보세요.

#### JavaScript: 브라우저의 작은 언어에서 웹의 공용 언어로

JavaScript의 출발점은 서버나 AI가 아니라 1995년 Netscape 브라우저였습니다. 당시 웹은 주로 정적인 HTML 문서였고, 버튼 반응이나 간단한 입력 확인처럼 페이지를 즉시 움직이게 할 언어가 필요했습니다. Netscape의 **Brendan Eich**가 만든 언어는 처음에 **Mocha**라고 불렸고, 잠시 **LiveScript**를 거쳐 **JavaScript**라는 이름을 얻었습니다. “10일 만에 설계됐다”는 유명한 일화는 이 언어가 완벽한 장기 설계보다 급변하던 웹 현장의 요구 속에서 빠르게 태어났다는 점을 보여 줍니다.

여기서 이름이 만드는 오해를 바로잡아야 합니다. **JavaScript는 Java의 가벼운 버전이 아닙니다.** 이름에 Java가 들어가지만, 두 언어는 별개의 언어입니다. Java는 클래스 기반·정적 타입 중심의 전통을 가졌고, JavaScript는 동적 타입과 프로토타입 기반 객체 모델을 사용합니다. 일부 문법 모양과 이름 관례가 비슷해 보일 수 있어도 설계 철학과 실행 방식이 같다는 뜻은 아닙니다. 이름을 마케팅의 흔적으로 보고, 실제 동작은 각각 따로 배우는 편이 안전합니다.

브라우저마다 JavaScript가 조금씩 다르게 동작하던 혼란을 줄이기 위해 언어 표준은 **ECMAScript**라는 이름으로 정리됐습니다. 우리가 일상적으로 JavaScript라고 부르는 언어의 표준 이름이 ECMAScript입니다. 특히 2015년의 **ES6(ECMAScript 2015)**는 `let`·`const`, 화살표 함수, 클래스 문법, 모듈, Promise 등 현대 문법을 널리 정착시킨 큰 전환점입니다. 그러므로 “JS를 배운다”는 말에는 브라우저에서 돌아가는 언어와 ECMAScript 표준의 변화가 함께 들어 있습니다.

#### Node.js: 브라우저 밖에서 JavaScript 실행하기

2009년 라이언 달(Ryan Dahl)이 만든 Node.js는 JavaScript를 내 컴퓨터와 서버에서도 실행하는 런타임입니다. React·Next.js 개발 도구와 프로젝트도 이 환경에서 실행합니다.

Node.js는 파일 읽기나 네트워크 요청처럼 기다리는 시간이 긴 일을 시작해 두고, 기다리는 동안 다른 일을 처리하는 방식에 강점이 있습니다. 이것이 “모든 작업에서 무조건 빠르다”는 뜻은 아닙니다. 큰 이미지 처리나 복잡한 계산처럼 CPU를 오래 점유하는 작업은 별도 전략이 필요합니다. 핵심은 기다리는 동안 다른 일을 처리하는 방식입니다.

#### npm: 패키지와 공급망

**npm(Node Package Manager)** 은 Node.js 생태계에서 필요한 코드 패키지를 설치하고, 프로젝트가 무엇에 의존하는지 기록하는 도구와 저장소입니다. 직접 로그인 화면, 날짜 처리, 개발 서버 기능을 모두 처음부터 만들지 않아도 되는 이유는 많은 사람이 만든 재사용 가능한 부품을 가져올 수 있기 때문입니다. 패키지는 재사용할 코드, `package.json`은 프로젝트가 어떤 부품을 쓰는지 적은 목록, 의존성은 내 앱이 작동하기 위해 기대는 부품입니다. `npm install`은 그 부품을 가져오고, `npm run dev`는 목록에 정의된 개발 작업을 실행합니다.

패키지를 설치하기 전에는 출처와 유지보수 상태를 확인해야 합니다. 2016년 작은 패키지 **left-pad**가 npm에서 삭제되자, 그 패키지를 간접적으로 사용하던 많은 프로젝트의 설치와 빌드가 영향을 받은 일이 있었습니다. 이 사건의 교훈은 “npm을 쓰지 말자”가 아닙니다. 외부 패키지도 업데이트·보안·유지 관리가 필요한 **공급망**이라는 점입니다. AI가 패키지를 추천해도 이름, 공식 문서, 최근 유지보수, 왜 필요한지를 한 번 확인하세요. `package.json`과 잠금 파일은 그 판단의 흔적이기도 합니다.

**오늘의 세 문장**

1. JavaScript는 웹페이지를 움직이기 위해 태어났지만, 이제 브라우저와 서버 양쪽에서 쓰이는 언어입니다.
2. Node.js는 JavaScript를 내 컴퓨터와 서버에서 실행하게 하는 런타임이며, Next.js 개발 도구를 돌리는 기반입니다.
3. npm은 필요한 코드 부품을 설치·기록하는 도구지만, 가져온 부품을 검토할 책임까지 대신하지는 않습니다.

### 개발 환경 설치

설치를 확인하기 전에, 필요한 도구가 같이 깔려 있어야 합니다. **Claude Code**(명령을 맡길 메인 도구), **Node.js와 npm**(웹 도구를 실행하는 환경), **Git**(변경 기록) 세 가지입니다. 이 중 Claude Code는 여러분이 이미 쓰고 있는 **Claude Desktop 안에서 바로 실행**할 수 있습니다.

두 길 중 하나를 고르세요. 어떤 길을 택하든, 마지막엔 **직접 확인하는 단계**로 이어집니다. "AI가 깔아줬다"에서 멈추지 않고, "내 컴퓨터에서 된다"를 눈으로 확인하는 것이 오늘의 목표입니다.

#### 빠른 길 - Claude Code에게 맡기기 (AI 네이티브)

Claude Desktop을 켜고, 그 안의 **Claude Code**를 엽니다. 그리고 자연어로 환경 세팅을 맡깁니다.

```text
이 컴퓨터에서 웹 개발을 시작하려고 해.
내가 macOS/Windows 환경이야. (본인 OS로 바꿔 말합니다)

확인해 줄 것:
1. Node.js와 npm이 깔려 있는지, 깔려 있지 않다면 공식 방법으로 설치해 줘.
2. Git이 깔려 있는지, 없다면 설치해 줘.
3. 각각 버전을 출력해 줘.

설치가 끝나면, 각 도구의 버전 명령(node --version, npm --version, git --version) 결과를 보여줘.
```

> 🤔 **왜 "버전을 보여줘"라고 덧붙일까요?** AI가 "설치했습니다"라고 답해도, 그 말만으로는 내 컴퓨터에서 진짜로 되는지 알 수 없습니다. 버전 출력은 **AI의 주장을 내가 검증**하는 가장 단순한 방법입니다. 이 습관이 앞으로 모든 바이브 코딩의 기본이 됩니다.

#### 손으로 직접 설치하기 (기본기)

도구가 어떻게 들어오는지 직접 보고 싶다면, 공식 문서의 방법으로 하나씩 깝니다. 공식 문서를 기준으로 삼는 이유는, 버전이 바뀌어도 가장 믿을 수 있는 출처이기 때문입니다.

- **Node.js와 npm - nvm으로 관리합니다**: Node.js 자체를 한 번 고정 설치하지 않고 **nvm(Node Version Manager)** 으로 설치·전환합니다. 프로젝트마다 필요한 Node.js 버전이 다를 수 있으므로, `nvm use`로 버전을 바꿀 수 있고 `--lts`로 현재 LTS를 관리할 수 있습니다. 즉 한 프로젝트의 Node.js가 다른 프로젝트를 덮어쓰지 않게 돕습니다. npm은 nvm으로 설치한 Node.js에 함께 들어옵니다.

  **권한 문제를 줄이는 이유도 있습니다.** macOS·Linux·WSL에서 Node.js를 시스템 위치에 직접 설치하면, 전역 패키지(`npm install -g ...`)를 설치할 때 쓰기 권한이 없어 `EACCES` 오류가 나고 `sudo`를 붙여야 하는 경우가 많습니다. 이 때문에 설치가 중단되는 학생도 많습니다.

  nvm은 기본적으로 홈 디렉터리의 `~/.nvm` 아래에서 Node.js 버전과 실행 바이너리 디렉터리를 관리합니다. 내 계정이 쓸 수 있는 위치를 사용하므로, nvm으로 설치한 Node.js의 전역 npm 패키지는 보통 `sudo` 없이 설치할 수 있어 권한 충돌을 피하는 데 도움이 됩니다. (`sudo npm install -g`는 사용하지 않습니다.)

  **macOS / Linux / WSL** — nvm 공식 저장소의 설치 스크립트를 실행합니다. 이 노트북 작성 시점의 공식 최신 릴리스는 **v0.40.8**입니다. 먼저 일반 터미널에서 설치 스크립트만 실행합니다.

  ```bash
  curl -o- https://raw.githubusercontent.com/nvm-sh/nvm/v0.40.8/install.sh | bash
  ```

  설치가 끝나면 터미널을 닫았다가 다시 열고 다음 명령으로 nvm이 로드됐는지 확인합니다. `nvm`이 출력된 뒤에 설치를 이어 갑니다. 출력이 없다면 [nvm 설치 안내](https://github.com/nvm-sh/nvm#installing-and-updating)의 셸 설정을 확인하세요.

  ```bash
  command -v nvm
  ```

  ```bash
  nvm install --lts
  nvm use --lts
  node --version
  npm --version
  ```

  `nvm install --lts`는 현재 LTS Node.js를 설치하고, `nvm use --lts`는 지금 이 터미널에서 그 LTS를 사용합니다. 다른 프로젝트가 특정 버전을 요구하면 그 프로젝트 폴더에서 예를 들어 `nvm install 22`와 `nvm use 22`처럼 전환합니다.

  **Windows (PowerShell)** — Unix용 nvm과 별개인 [nvm-windows 공식 Releases](https://github.com/coreybutler/nvm-windows/releases)에서 최신 `nvm-setup.exe`를 내려받아 설치합니다. 설치 뒤 새 터미널에서 다음처럼 LTS의 **명시적 버전 번호**를 설치·선택합니다(Windows용 nvm은 `--lts` 별칭을 제공하지 않습니다).

  ```powershell
  nvm install <LTS-버전>
  nvm use <LTS-버전>
  node --version
  npm --version
  ```

  `<LTS-버전>`은 [Node.js 공식 다운로드](https://nodejs.org/en/download) 페이지에 표시된 LTS 버전 번호로 바꿉니다. Windows에서는 관리자 권한이 필요한 경우가 있으므로, `nvm use`가 실패하면 관리자 권한으로 새 터미널을 열어 다시 실행합니다.

  > **출처 확인 습관**: 설치 명령을 복사하기 전 링크의 태그와 릴리스 페이지를 함께 확인합니다. nvm은 업데이트되므로, 수업 중 더 최신 공식 릴리스가 보이면 그 태그의 `install.sh` URL을 사용합니다.
- **uv (선택)**: [uv 공식 설치 문서](https://docs.astral.sh/uv/getting-started/installation/)의 설치 명령으로 깝니다. 지금 필요하지 않다면 건너뛰어도 됩니다. (macOS/Linux는 `curl -LsSf https://astral.sh/uv/install.sh | sh`, Windows는 공식 문서의 PowerShell 명령.)
- **Git**: [Git 공식 다운로드](https://git-scm.com/install/)에서 운영체제에 맞는 버전을 설치합니다.
- **Claude Code**: [Claude Code 공식 설치 문서](https://code.claude.com/docs/en/setup)를 기준으로 설치합니다. 터미널에서 `claude` 명령으로 불러옵니다.

#### 공통 버전 확인

두 길 모두, 설치가 끝났다고 넘어가지 말고 **직접 터미널을 열어 버전을 확인**합니다. 아래 명령의 출력으로 '작업대 상태표'를 채웁니다. CLI 대화 중이라면 `/exit`로 나와 실행하세요.

### AI에게 첫 파일 생성을 맡기는 프롬프트

백지에서 HTML을 외워 쓰지 않습니다. 먼저 AI에게 만들 것을 말하고, 나온 결과를 읽고 판단합니다. 프로젝트 폴더(`my-first-web`)에서 Claude Code를 실행한 뒤 아래 요청을 입력합니다. Claude Code가 `index.html`을 직접 만들어 줍니다. 대괄호 안은 자기 도메인으로 바꿉니다.

```text
index.html 파일에 브라우저에서 바로 열 수 있는 첫 화면을 만들어줘.
조건:
- 한국어 문서로 설정해줘.
- 큰 제목에는 [내 이름 또는 역할]을 넣어줘.
- 본문에는 [내 도메인]에서 AI를 활용해 보고 싶은 일을 한 문장으로 넣어줘.
```

이 요청은 코드를 대신 써 달라는 요청이지만, 책임까지 넘기는 요청은 아닙니다. 다음 셀에서 AI가 낸 결과 예시를 읽고 검토합니다.

```html
<!-- AI가 생성한 결과 예시입니다. 그대로 외워 쓰는 코드가 아니라 검토 대상입니다. -->
<!DOCTYPE html>
<html lang="ko">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>나의 첫 웹페이지</title>
</head>
<body>
  <h1>교육 운영 전문가 김아이펠</h1>
  <p>교육 현장에서 반복되는 안내 업무에 AI를 활용해 보고 싶습니다.</p>
</body>
</html>
```

## 바이브 코딩 루프 한 번 돌리기

---

이번 섹션이 오늘의 중심입니다. 설치와 파일 생성은 준비이고, 바이브 코딩은 여기서 시작됩니다.

바이브 코딩 루프는 네 단계입니다.

1. 만들 것을 자연어로 또렷이 말합니다.
2. AI가 코드를 생성합니다.
3. 여러분이 결과를 읽고 판단합니다.
4. 다시 지시해 수정합니다.

오늘은 작은 수정 하나로 이 루프를 경험합니다. 목표는 화려한 화면이 아니라, **AI 출력에 대해 근거 있는 재요청을 하는 것**입니다.

### 두 번째 요청: 도메인 섹션 추가

AI에게 아래처럼 다시 요청합니다. 첫 요청과 다른 점은 "무엇을 추가할지"와 "무엇은 건드리지 말지"를 함께 말한다는 점입니다.

## 터미널·셸·커널·프롬프트를 구분하기

## 터미널·셸·커널·프롬프트

명령을 입력하기 전에 네 역할만 구분해 봅니다.

| 이름 | 역할 |
|---|---|
| 터미널 | 키 입력을 전달하고 실행 결과를 보여 주는 창 |
| 셸 | 입력한 명령을 해석하고 프로그램을 실행하는 도구. macOS는 보통 zsh, 이 실습의 Windows는 PowerShell을 씁니다 |
| 커널 | 프로그램이 CPU 시간·메모리·파일·네트워크 같은 자원을 쓰게 중개하는 운영체제의 중심 |
| 프롬프트 | 셸이 명령을 받을 준비가 됐음을 알리는 표시 |

![터미널, PTY, 셸, 커널과 하드웨어의 계층도](https://ebhtmardjzsiyxjncvbb.supabase.co/storage/v1/object/public/images/01-vibe-coding-web/vibe-coding-web-1-fig04.svg)

CLI의 Claude Code는 셸에서 `claude`로 실행하는 프로그램입니다. 자연어는 Claude Code에, `pwd`·`ls` 같은 명령은 일반 터미널의 셸 프롬프트에 입력합니다.

**현재 작업 폴더는 상대 경로의 기준이지 보안 경계가 아닙니다.** 접근 범위는 운영체제 권한, Claude Code의 권한 설정, 추가 디렉터리 허용, 샌드박스 설정에 따라 달라집니다. `pwd`로 위치를 확인하고, 실제 파일 대신 새 연습 폴더 안의 파일만 다루세요. 이동 전에는 대상에 같은 이름의 파일이 있는지도 확인합니다.

아래 PTY·fork·exec와 관찰 명령은 Unix 계열(macOS·Linux) 기준입니다. Windows는 ConPTY·CreateProcess 등 다른 방식을 사용하므로 이 명령을 PowerShell에 그대로 입력하지 않습니다. 실습은 이 설명을 읽지 않아도 진행할 수 있습니다.

### 1. 이름의 역사: 끝, 껍데기, 알맹이, 재촉

**터미널(terminal)** 은 원래 검은 창의 이름이 아니었습니다. 통신과 전기 분야에서 terminal은 선로의 **끝단**, 곧 중앙 시스템과 사람이 만나는 말단 장치를 뜻했습니다. 1960～70년대에는 값비싼 대형 컴퓨터 한 대를 여러 사람이 함께 썼습니다. 사람들은 본체 앞에 앉는 대신 키보드와 화면, 또는 종이를 찍는 텔레타이프 장치로 중앙 컴퓨터에 연결했습니다. 그 물리 장치가 터미널입니다. 오늘날 macOS Terminal, iTerm2, Windows Terminal, VS Code 내장 터미널은 옛 장치의 입출력 방식을 화면 속에서 재현하는 **터미널 에뮬레이터**입니다. 창을 열고 글자를 보여 주며 키 입력을 전달하지만, 보통 명령의 뜻을 해석하지는 않습니다.

**셸(shell)** 은 껍데기입니다. 운영체제의 핵심인 커널을 감싼 바깥층이라는 비유에서 온 이름입니다. Unix에는 1971년 Thompson shell, 1977년 Stephen Bourne의 Bourne shell(`/bin/sh`), 1989년 Brian Fox의 bash(Bourne Again Shell)가 이어졌고, macOS에서는 zsh도 널리 쓰입니다. 셸은 한 종류의 앱 이름이 아니라 명령을 해석하는 프로그램의 계열입니다.

**커널(kernel)** 은 ‘알맹이, 핵심’이라는 뜻으로, 프로그램이 CPU 시간, 메모리, 파일, 디스크, 네트워크 같은 자원을 쓰게 중개하는 운영체제의 중심입니다. 우리는 보통 커널에 직접 문장을 쓰지 않습니다. 대신 셸이 사람이 쓴 글을 해석하여 프로그램을 실행하고, 그 프로그램과 커널이 만나는 길을 마련합니다.

**프롬프트(prompt)** 는 ‘재촉하다, 응답을 요구하다’라는 뜻입니다. 셸이 “입력을 기다리고 있습니다”라고 화면에 내보내는 표시가 프롬프트입니다. `$`, `%`, `user@computer:~$`처럼 보일 수 있습니다. 프롬프트는 명령어가 아니므로 복사해서 입력하는 대상이 아닙니다. 예를 들어 문서에 `% pwd`라고 적혀 있다면 `%`는 예시 프롬프트이고, 실제로 입력할 것은 `pwd`입니다. 프롬프트의 모양은 셸 설정에 따라 바뀌지만, 입력 대기 상태라는 역할은 같습니다.

그림은 층을 단순화한 지도입니다. 터미널과 셸은 같은 것이 아니며, 프롬프트는 셸이 터미널에 그려 달라고 한 작은 안내문입니다. 그 사이에 **PTY(pseudoterminal, 의사 터미널)** 가 있습니다. 현대 운영체제에서 터미널 에뮬레이터와 셸은 직접 손을 잡지 않고, master와 slave로 짝을 이룬 가상 문자 장치를 통해 글자를 주고받습니다. 터미널 앱은 보통 master 쪽, 셸은 옛 물리 터미널처럼 보이는 slave 쪽을 사용합니다. 이 구조 덕분에 창 속 프로그램은 여전히 전통적인 터미널에 연결된 것처럼 동작합니다.

### 2. 작동 메커니즘: `% pwd` 한 줄이 거치는 일곱 단계

이제 프롬프트 뒤에 `pwd`를 입력하고 Enter를 눌렀다고 해 봅시다. `%`는 입력하지 않는 프롬프트라는 점을 다시 기억하세요. 화면에서 일어나는 일은 대략 다음 일곱 단계입니다.

1. **터미널이 키 입력을 받는다.** 터미널 에뮬레이터는 `p`, `w`, `d`, Enter 키를 받아 화면에 글자를 보이고 PTY의 master 쪽으로 전달합니다. 이 단계는 ‘창’의 일입니다.
2. **PTY가 셸에 글자를 넘긴다.** 가상 터미널의 slave 쪽에 연결된 셸이 Enter까지 모인 입력을 읽습니다. 화면에 명령이 보인다고 해서 아직 프로그램이 실행된 것은 아닙니다.
3. **셸이 문장을 파싱한다.** zsh나 bash는 공백, 따옴표, 변수, 파이프(`|`), 리디렉션(`>`) 같은 문법을 해석합니다. `pwd`는 단순하지만, `echo "$HOME" | grep user` 같은 문장은 셸의 해석이 특히 중요합니다.
4. **셸이 명령의 위치를 찾는다.** 외부 프로그램이라면 셸은 `$PATH`에 적힌 디렉터리를 앞에서부터 살펴 실행 파일을 찾습니다. 다만 `pwd`는 많은 셸에서 **내장 명령(builtin)** 이기도 합니다. 이 경우 셸은 굳이 디스크에서 `/bin/pwd`를 찾지 않고 자기 기능으로 현재 경로를 출력할 수 있습니다. 그래서 `which pwd`의 결과와 실제 실행 방식이 언제나 완전히 같다고 단정하면 안 됩니다.
5. **필요하면 fork와 exec로 새 프로그램을 시작한다.** 외부 명령을 실행할 때 Unix 계열 셸은 보통 자기 프로세스를 복제하는 `fork` 뒤, 새 프로세스에 실행 파일을 올리는 `exec`를 사용합니다. 셸은 다음 명령을 받기 위해 남고, 새 프로그램은 독립적으로 일합니다. 내장 명령은 셸 안에서 바로 처리될 수도 있습니다.
6. **프로그램이 시스템 콜로 커널에 요청한다.** 파일을 읽거나, 현재 디렉터리를 확인하거나, 네트워크에 접속하는 일은 프로그램이 운영체제에 요청해야 합니다. 시스템 콜은 사용자 프로그램이 커널 기능을 안전하게 쓰는 문입니다. 커널은 권한을 확인하고 CPU·메모리·파일 시스템 같은 자원을 중개합니다.
7. **stdout이 되돌아와 화면에 보인다.** 프로그램이 표준 출력(stdout)에 결과를 쓰면, 그 글자는 PTY를 거쳐 터미널 에뮬레이터로 돌아옵니다. 터미널은 글자를 화면에 그린 뒤, 셸이 새 프롬프트를 표시할 수 있게 합니다. `pwd`의 경로 한 줄은 이렇게 돌아온 결과입니다.

이 일곱 단계는 모든 명령이 똑같이 복제된다는 뜻은 아닙니다. 내장 명령인지, 별도 프로그램인지, 파이프나 리디렉션이 있는지에 따라 세부는 달라집니다. 그러나 “창이 곧 명령어”라는 생각보다 훨씬 정확한 기본 모형입니다. 오류도 이 지도로 나눠 볼 수 있습니다. 창이 열리지 않으면 터미널 앱 문제일 수 있고, `command not found`는 셸의 명령 탐색 또는 설치 문제일 수 있으며, `Permission denied`는 커널이 관리하는 권한 규칙과 관련될 수 있습니다.

### 3. 흔한 오해 바로잡기: 검은 창 하나가 전부가 아니다

가장 흔한 오해는 **“터미널 = 검은 창 = 명령어”** 입니다. 검은색은 꾸미기일 뿐입니다. 밝은 테마의 Windows Terminal도 터미널이고, VS Code 아래쪽 패널도 터미널이 될 수 있습니다. 터미널은 글자의 입출력 인터페이스, 셸은 그 글자로 된 명령을 해석하는 프로그램, 명령은 실제로 실행되는 내장 기능 또는 별도 프로그램입니다. 세 층을 분리해 말하면 문제를 더 빨리 설명할 수 있습니다.

터미널과 셸은 서로 바꿀 수 있습니다. 같은 macOS Terminal 창에서도 `bash`를 실행하면 bash 세션을 시작할 수 있고, 기본 로그인 셸을 바꾸는 설정도 있습니다. 반대로 같은 zsh를 iTerm2, macOS Terminal, VS Code 내장 터미널에서 쓸 수 있습니다. 따라서 “iTerm2를 설치했으니 bash가 설치됐다”거나 “VS Code를 열었으니 셸이 바뀌었다”는 결론은 맞지 않습니다. 예를 들어 **iTerm2(터미널) + zsh(셸)**, **VS Code 내장 터미널(터미널) + bash(셸)** 조합이 모두 가능합니다.

또 하나의 오해는 프롬프트 기호를 명령에 포함하는 것입니다. 수업 자료의 `$ npm --version`에서 `$`는 설명을 위한 표시일 수 있습니다. 복사할 때는 `npm --version`만 입력합니다. 프롬프트는 설정으로 사용자 이름, 현재 폴더, Git 브랜치까지 보여 주도록 바꿀 수 있으므로 사람마다 다르게 보이는 것이 정상입니다.

### 4. 실습 관찰: 내 작업대의 각 층을 직접 찾기

아래 명령은 컴퓨터를 바꾸지 않고 정보를 출력합니다. 한 줄씩 실행한 뒤, 무엇을 알려 주는지 결과 옆에 적어 보세요.

```bash
echo $SHELL
echo $0
tty
which pwd
echo $PATH
```

> 🪟 **Windows PowerShell 사용자**: 위 관찰 명령은 macOS/Linux 셸(zsh·bash) 기준이라 `echo $SHELL`, `echo $0`, `tty`, `which`는 그대로 동작하지 않습니다. 대신 `$PSVersionTable`(지금 셸이 PowerShell임을 확인), `Get-Command pwd`(명령을 어디서 찾았는지), `$env:Path`(명령 탐색 경로 목록)를 관찰하세요. 관찰 포인트는 같습니다 — 지금 내 셸이 무엇이고, 명령을 어디서 찾는가.

- `echo $SHELL`은 **로그인 셸로 설정된 경로**를 보여 줍니다. macOS에서는 `/bin/zsh` 같은 값이 흔합니다. 단, 이것이 지금 이 순간 실행 중인 셸과 반드시 같지는 않습니다. 다른 셸을 실행해 들어왔다면 설정과 현재 상태가 달라질 수 있습니다.
- `echo $0`은 현재 실행 중인 셸 이름을 관찰하는 데 쓸 수 있습니다. 환경과 실행 방법에 따라 앞에 `-`가 붙는 등 모양이 달라질 수 있으므로, `echo $SHELL`과 결과가 다르다고 곧바로 오류라고 판단하지 마세요.
- `tty`는 현재 세션이 연결된 터미널 장치 경로를 보여 줍니다. 예를 들어 `/dev/ttys001`처럼 보일 수 있습니다. 이것이 터미널·PTY의 흔적입니다. 그래픽 창의 이름이 아니라 운영체제가 보는 문자 장치를 확인하는 명령입니다.
- `which pwd`는 `$PATH`를 따라 발견한 `pwd`의 위치를 알려 줍니다. 다만 앞에서 본 것처럼 `pwd`는 셸 내장 명령일 수 있어, 더 자세히 알고 싶다면 `type pwd`도 관찰해 보세요. `which`의 한 줄만으로 실행 방식을 모두 설명할 수는 없습니다.
- `echo $PATH`는 셸이 외부 명령을 찾을 때 순서대로 보는 디렉터리 목록입니다. 콜론(`:`)으로 구분된 여러 경로가 보입니다. Node.js나 Claude Code를 설치했는데 `command not found`가 나오면, 설치 실패뿐 아니라 실행 파일이 이 목록에 없는 경우도 살펴볼 수 있습니다.

**관찰 질문**: `echo $SHELL`과 `echo $0`은 같은가요? `tty`는 어떤 `/dev/...` 값을 보여 주나요? `$PATH`에 Node 또는 nvm 관련 경로가 있나요? 답이 다르더라도 정상일 수 있습니다. 출력 원문과 “무엇을 기대했는지”를 함께 AI에게 보여 주면, AI가 추측으로 명령을 제시하는 일을 줄일 수 있습니다.

### 5. Claude Code는 어디에 있는가

Claude Code는 터미널도, 셸도, 커널도 아닙니다. 일반적으로 **셸 위에서 실행되는 프로그램**입니다. 터미널 창에서 `claude`를 입력하면 셸이 `$PATH`에서 `claude`를 찾고, 필요하면 새 프로세스를 실행합니다. Claude Code는 그 셸 환경과 권한 안에서 파일을 읽고, 명령 실행을 요청하고, 결과와 오류를 다시 보여 줍니다. 그러므로 AI가 “명령을 실행했다”고 말할 때도, 실제로는 이 여러 층을 지나는 것입니다.

이 구분은 협업에서 특히 유용합니다. `claude: command not found`라면 먼저 설치 또는 `$PATH`처럼 셸의 명령 탐색을 의심합니다. `EACCES`나 `Permission denied`라면 운영체제가 관리하는 파일 권한을 확인합니다. 글자가 깨지거나 창 자체가 멈추면 터미널 앱·인코딩·PTY 연결을 살펴봅니다. 현재 작업 폴더는 상대 경로의 기준이지 보안 경계가 아닙니다. 접근 범위는 운영체제 권한, Claude Code의 권한 설정, 추가 디렉터리 허용, 샌드박스 설정에 따라 달라집니다. 따라서 오류 한 줄, `pwd`, `echo $SHELL`, 실행한 명령을 함께 주는 습관은 AI에게 더 안전하고 재현 가능한 도움을 요청하는 방법입니다.

## 실습 1 — 터미널로 폴더를 탐험하고 만들기

---

### 안전한 연습 폴더에서 시작하기

> **안전 규칙**: 아래 명령은 삭제 명령을 쓰지 않습니다. 바탕화면의 실제 파일 대신 반드시 새로 만든 `terminal-practice` 연습 폴더에서 실행하세요. 공백이 있는 이름은 따옴표로 감쌉니다.

| 명령 | 하는 일 |
|---|---|
| `pwd` | 지금 내가 있는 폴더의 전체 경로를 출력합니다. |
| `ls` | 현재 폴더 안의 파일·폴더 이름을 보여 줍니다. |
| `cd 폴더이름` | 지정한 폴더로 이동합니다. `cd ..`는 한 단계 위로 갑니다. |
| `mkdir 폴더이름` | 새 폴더를 만듭니다. |
| `touch 파일이름` | 빈 파일을 만들거나, 기존 파일의 수정 시각을 갱신합니다. |
| `mv 원본 대상` | 파일·폴더를 이동하거나 이름을 바꿉니다. 대상에 같은 이름이 있으면 덮어쓸 수 있으므로 먼저 확인합니다. |

명령은 **일반 터미널**에 입력합니다. Claude Code CLI 대화 중이라면 `/exit`로 나옵니다. Windows는 **PowerShell**을 사용합니다.

#### 1단계 — 내 위치를 보고 새 연습 공간 만들기

일반 터미널에서 `pwd`와 `ls`로 위치와 목록을 확인하세요. `terminal-practice`가 이미 있으면 들어가지 말고 `terminal-practice-2`처럼 겹치지 않는 이름으로 바꿉니다. 각 줄이 성공한 뒤 다음 줄을 실행하세요.

**macOS 터미널**

```bash
mkdir terminal-practice
cd terminal-practice
mkdir ideas assets archive
touch ideas/landing-page.txt sample.png sample.jpg notes.txt
ls
```

**Windows PowerShell**

```powershell
mkdir terminal-practice
cd terminal-practice
mkdir ideas, assets, archive
New-Item ideas/landing-page.txt, sample.png, sample.jpg, notes.txt -ItemType File
ls
```

`.png`·`.jpg` 파일은 확장자 분류만 연습할 **빈 가짜 파일**입니다. 실제 이미지가 아니므로 열어 볼 필요는 없습니다. `pwd` 출력을 적고 세 폴더와 가짜 파일이 있는지 확인하세요.

#### 2단계 — 파일을 직접 이동하기

아래 명령은 두 환경에서 같습니다. `assets/images`가 새 빈 폴더인지 확인한 뒤 이동하세요. 같은 이름의 대상 파일이 보이면 이동하지 않습니다.

```bash
mkdir assets/images
ls assets/images
mv ideas/landing-page.txt assets/images/landing-page.txt
ls assets/images
pwd
```

빈 메모 파일 하나를 옮겼습니다. 루트의 `sample.png`, `sample.jpg`, `notes.txt`는 AI 정리 실습에 사용합니다.

#### 3단계 — 길을 잃었을 때 돌아오기

```bash
cd assets
pwd
cd ..
pwd
ls
```

`cd ..`는 “상위 폴더”로 가는 명령입니다. 마지막 `pwd`가 다시 `terminal-practice`를 가리키면 성공입니다. 화면에 보이는 이름만 믿지 말고, 이동할 때마다 `pwd`와 `ls`를 번갈아 사용하세요. **실습 2도 이 폴더에서 이어 가므로 여기서 더 나가지 않습니다.**

> **멈춤 신호**: 현재 위치가 불안하면 `pwd`부터 실행합니다. `mv`는 대상 경로를 한 글자씩 확인한 뒤 실행합니다.

## AI와 연습 폴더를 안전하게 정리하기

## 실습 2 — AI에게 폴더 정리를 맡기고 검토하기

---

### AI와 연습 폴더 정리하기

실습 1에서 만든 **같은 `terminal-practice` 폴더의 가짜 파일만** 정리합니다. 실제 바탕화면·문서·다운로드 폴더는 대상으로 삼지 않습니다. Desktop에서는 이 연습 폴더를 작업 폴더로 선택하고, CLI에서는 `pwd`로 위치를 확인한 뒤 `claude`를 실행합니다. 다른 이름으로 만들었다면 그 이름을 사용하세요.

아래 자연어 요청은 Claude Code에 입력합니다. 작업 폴더 선택만으로 외부 파일 접근이 차단되지는 않으므로, 계획의 전체 경로를 확인한 뒤 승인합니다.

#### Claude Code에 보낼 프롬프트 예시

```text
현재 작업 폴더의 절대 경로를 먼저 보여줘. 실습 1에서 만든 연습 폴더가 맞는지 내가 확인한 뒤, 그 안의 가짜 파일만 읽고 정리 계획을 세워줘. 폴더 밖 경로나 심볼릭 링크는 따라가지 마.

규칙:
1. 삭제 명령(rm 등)은 절대 실행하지 마.
2. 먼저 dry-run으로 파일별 '현재 위치 → 제안 위치' 목록과 분류 기준만 보여줘.
3. 이동은 내가 목록을 검토하고 '승인'이라고 답한 뒤에만 실행해.
4. 이름을 추측하지 말고, 확장자와 현재 이름을 근거로 분류해.
5. 이동 대상에 같은 이름이 있으면 덮어쓰거나 이름을 바꾸지 말고 중단해서 물어봐.
6. 실행 뒤에는 이동한 목록과 남은 파일 목록을 보여줘.
```

`dry-run`은 실제 변경 없이 “무엇을 할지” 미리 보여 주는 실행 계획입니다. AI가 먼저 `mv`를 실행하려 하면, “아직 실행하지 말고 dry-run 목록만 보여줘”라고 바로 중단시킵니다.

#### AI 산출물 검토 체크리스트

| 확인할 것 | 통과 기준 | 내 판단 |
|---|---|---|
| 파일 이동 전 dry-run | 각 파일의 현재 위치와 제안 위치가 목록으로 있다 | ☐ |
| 덮어쓰기 방지 | 대상에 같은 이름이 없는지 확인하고, 있으면 이동을 멈춘다 | ☐ |
| 삭제 방지 | 삭제 명령·삭제 제안이 없고, 보류 파일은 그대로 둔다 | ☐ |
| 분류 기준 | 확장자·용도 등 읽을 수 있는 기준이 있고 애매한 파일은 질문한다 | ☐ |
| 경로 안전 | 작업 폴더 밖의 파일을 건드리지 않고, 중요한 폴더를 통째로 이동하지 않는다 | ☐ |
| 실행 후 확인 | `ls` 또는 목록으로 이동 결과와 남은 항목을 보여 준다 | ☐ |

#### 승인·수정 요청 예시

```text
계획과 대상의 기존 파일을 확인했어. sample.png와 sample.jpg만 assets/images 폴더로 옮기고,
확장자를 알 수 없는 파일과 프로젝트 폴더는 그대로 둬.
삭제·이름 변경·덮어쓰기는 하지 마. 대상에 같은 이름이 있으면 멈춰. 실행 후 이동 전후 목록을 비교해 보여줘.
```
